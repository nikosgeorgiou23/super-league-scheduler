# 1. Εισαγωγή

### 1.1. Αθλητικός Προγραμματισμός και Τηλεοπτικές Απαιτήσεις

Ο προγραμματισμός των αγώνων ενός επαγγελματικού πρωταθλήματος ποδοσφαίρου αποτελεί ένα σύνθετο πρόβλημα συνδυαστικής βελτιστοποίησης, καθώς απαιτεί τον συντονισμό πολλών, συχνά αντικρουόμενων, απαιτήσεων. Η διαδικασία δεν περιορίζεται στον καθορισμό των αναμετρήσεων κάθε αγωνιστικής, αλλά περιλαμβάνει επίσης την ανάθεση των αγώνων σε συγκεκριμένες ημερομηνίες και χρονικές ζώνες, λαμβάνοντας υπόψη αγωνιστικούς, οργανωτικούς και εμπορικούς παράγοντες. Ιδιαίτερη σημασία έχουν οι τηλεοπτικές απαιτήσεις, καθώς η επιλογή της ημέρας και της ώρας διεξαγωγής ενός αγώνα μπορεί να επηρεάσει τη δυνητική τηλεθέαση και την εμπορική αξία της μετάδοσής του. Οι προτιμήσεις των τηλεοπτικών παρόχων ενδέχεται να εξαρτώνται από τη δημοτικότητα των ομάδων, την ελκυστικότητα των αναμετρήσεων και την κατανομή των αγώνων στις διαθέσιμες χρονικές ζώνες. Η ανάγκη ταυτόχρονης ικανοποίησης υποχρεωτικών περιορισμών και επιδίωξης πολλαπλών, ενδεχομένως ανταγωνιστικών, στόχων καθιστά τον προγραμματισμό ενός ποδοσφαιρικού πρωταθλήματος ιδιαίτερα απαιτητικό, αναδεικνύοντας τη δυνατότητα των μεθόδων μαθηματικής βελτιστοποίησης να προσφέρουν συστηματικές και τεκμηριωμένες λύσεις.

### 1.2. Προκλήσεις Προγραμματισμού στην Ελληνική Super League

Η ελληνική Super League παρουσιάζει ιδιαίτερες προκλήσεις στον προγραμματισμό των αγώνων, καθώς τα ζητήματα δημόσιας ασφάλειας και τα περιστατικά οπαδικής βίας εισάγουν πρόσθετη πολυπλοκότητα, πέρα από τις αγωνιστικές, οργανωτικές και εμπορικές απαιτήσεις που συναντώνται συνήθως στις επαγγελματικές ποδοσφαιρικές διοργανώσεις. Οι έντονες αντιπαλότητες μεταξύ συγκεκριμένων συλλόγων, τα προηγούμενα περιστατικά βίας και η πιθανότητα συγκρούσεων μεταξύ ομάδων οπαδών δημιουργούν αυξημένες απαιτήσεις για τη διαχείριση αναμετρήσεων υψηλής επικινδυνότητας. Ιδιαίτερη προσοχή πρέπει να δίνεται στη γεωγραφική εγγύτητα των γηπέδων, στον ταυτόχρονο προγραμματισμό αγώνων που ενδέχεται να προκαλέσουν ζητήματα ασφάλειας, καθώς και στη διαθεσιμότητα αστυνομικών δυνάμεων για την αποτελεσματική επιτήρηση των αθλητικών εκδηλώσεων. Η ενσωμάτωση τέτοιων απαιτήσεων σε ένα μαθηματικό μοντέλο βελτιστοποίησης μπορεί, επομένως, να συμβάλει στη δημιουργία προγραμμάτων που λαμβάνουν συστηματικά υπόψη τους παράγοντες που σχετίζονται με την ασφάλεια.

### 1.3. Συνδυαστική και Υπολογιστική Πολυπλοκότητα

Η μαθηματική δυσκολία του προγραμματισμού ενός ποδοσφαιρικού πρωταθλήματος προκύπτει τόσο από το μεγάλο πλήθος πιθανών χρονικών αναθέσεων όσο και από τις αλληλεξαρτήσεις μεταξύ των περιορισμών προγραμματισμού. Έστω $M$ το σύνολο των αγώνων μιας διοργάνωσης και $S_m$ το σύνολο των διαθέσιμων χρονικών ζωνών για κάθε αγώνα $m$. Υποθέτοντας ότι κάθε αγώνας μπορεί αρχικά να ανατεθεί ανεξάρτητα σε οποιαδήποτε από τις διαθέσιμες χρονικές ζώνες του, το συνολικό πλήθος των πιθανών αναθέσεων, πριν από την επιβολή πρόσθετων περιορισμών, δίνεται από τη σχέση:

$$
|\Omega_0| = \prod_{m \in M} |S_m|
$$

Ενδεικτικά, θεωρούμε ένα πρωτάθλημα που αποτελείται από 14 ομάδες, 26 αγωνιστικές και 182 αγώνες. Εάν κάθε αγώνας μπορεί να ανατεθεί σε μία από οκτώ διαθέσιμες χρονικές ζώνες, το θεωρητικό πλήθος των πιθανών αναθέσεων είναι:

$$
|\Omega_0| = 8^{182} \approx 2.30 \times 10^{164}
$$

Η τιμή αυτή αντιπροσωπεύει το μέγεθος του χώρου αναζήτησης χωρίς περιορισμούς και όχι το πλήθος των εφικτών προγραμμάτων.

Στην πράξη, οι αγωνιστικοί, οργανωτικοί, τηλεοπτικοί περιορισμοί, καθώς και οι απαιτήσεις δημόσιας ασφάλειας, δημιουργούν αλληλεξαρτήσεις που περιορίζουν σημαντικά τις διαθέσιμες αναθέσεις και καθιστούν πιο σύνθετη την αναζήτηση εφικτών λύσεων. Επιπλέον, ο εντοπισμός ενός εφικτού προγράμματος δεν εγγυάται απαραίτητα και τη βέλτιστη ποιότητά του, γεγονός που αναδεικνύει τη σημασία της αξιοποίησης εξειδικευμένων μεθόδων μαθηματικής βελτιστοποίησης.

### 1.4. Αντικείμενο της Εργασίας και Ερευνητική Συνεισφορά

Στην πράξη, ο προγραμματισμός των επαγγελματικών ποδοσφαιρικών πρωταθλημάτων πραγματοποιείται συχνά σε διαδοχικά στάδια. Αρχικά καθορίζονται οι αναμετρήσεις και η κατανομή τους στις αγωνιστικές, ενώ η διαδικασία του αναλυτικού χρονικού προγραμματισμού, που περιλαμβάνει την ανάθεση των αγώνων σε συγκεκριμένες ημερομηνίες και ώρες έναρξης, μπορεί να πραγματοποιηθεί σε μεταγενέστερο στάδιο, ανάλογα με τις διαθέσιμες πληροφορίες και τις λειτουργικές απαιτήσεις. Η προσέγγιση αυτή, η οποία παρατηρείται και στην ελληνική Super League, επιτρέπει τη σταδιακή λήψη αποφάσεων προγραμματισμού και την προσαρμογή τους σε μεταβαλλόμενες συνθήκες, όπως οι υποχρεώσεις των ομάδων στις ευρωπαϊκές διοργανώσεις, οι τηλεοπτικές προτεραιότητες και οι οργανωτικοί περιορισμοί.

Αντίθετα, η παρούσα εργασία εξετάζει μια προσέγγιση χρονικού προγραμματισμού σε επίπεδο ολόκληρης της αγωνιστικής περιόδου, στην οποία οι αναμετρήσεις και η κατανομή τους στις αγωνιστικές θεωρούνται προκαθορισμένες, ενώ οι ημερομηνίες και οι ώρες έναρξης όλων των αγώνων της κανονικής διάρκειας του πρωταθλήματος βελτιστοποιούνται στο πλαίσιο ενός ενιαίου μαθηματικού μοντέλου. Με τον τρόπο αυτό, καθίσταται δυνατή η ταυτόχρονη εξέταση απαιτήσεων προγραμματισμού που εκτείνονται σε ολόκληρη την αγωνιστική περίοδο, αντί της ανεξάρτητης αντιμετώπισής τους για κάθε αγωνιστική.

Πρωταρχικός στόχος είναι η κατασκευή ενός εφικτού προγράμματος που ικανοποιεί όλους τους υποχρεωτικούς περιορισμούς, επιδιώκοντας παράλληλα την ισορροπημένη κατανομή της τηλεοπτικής ελκυστικότητας μεταξύ των αγωνιστικών. Μια τέτοια προσέγγιση είναι ιδιαίτερα κατάλληλη για διοργανώσεις με σχετικά σταθερό αγωνιστικό ημερολόγιο και περιορισμένες εξωτερικές υποχρεώσεις, συμπεριλαμβανομένων πρωταθλημάτων των οποίων οι συμμετέχουσες ομάδες δεν αγωνίζονται σε διεθνείς διασυλλογικές διοργανώσεις.

Το προτεινόμενο μοντέλο αποσκοπεί στην ανάπτυξη μιας διαφανούς και συστηματικής μεθοδολογίας για τον χρονικό προγραμματισμό ολόκληρης της αγωνιστικής περιόδου, με δυνατότητα μελλοντικής προσαρμογής σε πιο σύνθετα αγωνιστικά περιβάλλοντα. Η υλοποίησή του βασίζεται σε ρητά καθορισμένες παραδοχές σχετικά με τη διαθεσιμότητα των δεδομένων και τις συνθήκες διεξαγωγής του πρωταθλήματος, οι οποίες παρουσιάζονται στο επόμενο κεφάλαιο, όπου διατυπώνεται αναλυτικά το πρόβλημα.


# 2. Διατύπωση του Προβλήματος

### 2.1. Δομή του Πρωταθλήματος

- Στο πρωτάθλημα συμμετέχουν 14 ομάδες και η κανονική διάρκεια αποτελείται από δύο γύρους των 13 αγωνιστικών:
  - **Πρώτος γύρος:** Αγωνιστικές 1–13.
  - **Δεύτερος γύρος:** Αγωνιστικές 14–26.


**Μαθηματική Διατύπωση**

Έστω $T$ το σύνολο των ομάδων και $R$ το σύνολο των αγωνιστικών.

$$
|T|=14
$$

$$
R=\{1,2,\ldots,26\}
$$

Ορίζουμε τα σύνολα αγωνιστικών των δύο γύρων:

$$
R_1=\{1,2,\ldots,13\}
$$

$$
R_2=\{14,15,\ldots,26\}
$$

με:

$$
R=R_1\cup R_2,\qquad R_1\cap R_2=\varnothing
$$

Το σύνολο $T$ περιλαμβάνει τις ακόλουθες 14 ομάδες:

- Παναθηναϊκός
- Ολυμπιακός
- ΑΕΚ
- ΠΑΟΚ
- Άρης
- Ηρακλής
- Ατρόμητος
- Κηφισιά
- Καλαμάτα
- Αστέρας Τρίπολης
- Παναιτωλικός
- ΟΦΗ
- Λεβαδειακός
- Βόλος

Επιπλέον, ορίζουμε το σύνολο $M$ των αναμετρήσεων του πρωταθλήματος:

$$
M=\{(i,j)\in T\times T\mid i\neq j\}
$$

Κάθε διατεταγμένο ζεύγος $(i,j)\in M$ αντιπροσωπεύει μία αναμέτρηση στην οποία η ομάδα $i$ είναι γηπεδούχος και η ομάδα $j$ φιλοξενούμενη. Επομένως, οι αναμετρήσεις $(i,j)$ και $(j,i)$ θεωρούνται διαφορετικές, καθώς αντιστοιχούν σε αγώνες με αντίστροφες έδρες.



- Κάθε ομάδα συμμετέχει σε ακριβώς μία αναμέτρηση ανά αγωνιστική, είτε ως γηπεδούχος είτε ως φιλοξενούμενη.

**Μαθηματική Διατύπωση**

Ορίζουμε τη δυαδική μεταβλητή απόφασης:

$$
x_{ijr}=
\begin{cases}
1, & \text{αν η ομάδα }i\text{ υποδέχεται την }j\text{ στην αγωνιστική }r\\
0, & \text{διαφορετικά}
\end{cases}
$$

για $i,j\in T$, $i\neq j$ και $r\in R$.

Ο περιορισμός συμμετοχής κάθε ομάδας είναι:

$$
\sum_{\substack{j\in T\\j\neq i}}
(x_{ijr}+x_{jir})=1
\qquad \forall i\in T,\ \forall r\in R
$$



- Κάθε ζευγάρι ομάδων αναμετράται ακριβώς μία φορά σε κάθε γύρο, με αντίστροφες έδρες μεταξύ πρώτου και δεύτερου γύρου. Η σειρά των αναμετρήσεων στον δεύτερο γύρο δεν είναι υποχρεωτικά ίδια με εκείνη του πρώτου.

**Μαθηματική Διατύπωση**

Για κάθε ζευγάρι διαφορετικών ομάδων $i,j\in T$, με $i<j$, επιβάλλεται μία αναμέτρηση ανά γύρο:

$$
\sum_{r\in R_1}(x_{ijr}+x_{jir})=1
\qquad \forall i,j\in T,\ i<j
$$

$$
\sum_{r\in R_2}(x_{ijr}+x_{jir})=1
\qquad \forall i,j\in T,\ i<j
$$

Για να εξασφαλιστεί η αντιστροφή των εδρών μεταξύ των δύο γύρων:

$$
\sum_{r\in R_1}x_{ijr}
=
\sum_{r\in R_2}x_{jir}
\qquad \forall i,j\in T,\ i\neq j
$$



**Παράγωγες Ιδιότητες**

Από τους παραπάνω περιορισμούς προκύπτουν οι ακόλουθες ιδιότητες:

- Κάθε ομάδα δίνει συνολικά 13 εντός έδρας και 13 εκτός έδρας αγώνες.
- Κάθε αγωνιστική περιλαμβάνει ακριβώς 7 αναμετρήσεις.
- Στην κανονική διάρκεια του πρωταθλήματος διεξάγονται συνολικά 182 αγώνες.

**Μαθηματική Διατύπωση**

Το πλήθος των εντός και εκτός έδρας αναμετρήσεων κάθε ομάδας είναι:

$$
H_i=A_i=|T|-1=13,\qquad \forall i\in T
$$

όπου $H_i$ και $A_i$ συμβολίζουν το πλήθος των εντός και εκτός έδρας αγώνων της ομάδας $i$, αντίστοιχα.

Το πλήθος των αναμετρήσεων ανά αγωνιστική είναι:

$$
\frac{|T|}{2}=\frac{14}{2}=7
$$

Το συνολικό πλήθος των αναμετρήσεων του πρωταθλήματος είναι:

$$
|M|=|T|(|T|-1)=14\cdot13=182
$$

Οι παραπάνω σχέσεις αποτελούν συνέπειες των βασικών περιορισμών και δεν απαιτείται να επιβληθούν ως πρόσθετοι ανεξάρτητοι περιορισμοί στο μαθηματικό μοντέλο.



### 2.2. Χαρακτηριστικά Ομάδων

**Κατηγοριοποίηση Ομάδων Υψηλού Κινδύνου**

- Για τις ανάγκες του προγραμματισμού, ορισμένες ομάδες χαρακτηρίζονται ως υψηλού κινδύνου και διαχωρίζονται σε δύο κατηγορίες με βάση τη γεωγραφική τους τοποθεσία:
  - **Ομάδες υψηλού κινδύνου Αθήνας:** Παναθηναϊκός, Ολυμπιακός, ΑΕΚ.
  - **Ομάδες υψηλού κινδύνου Θεσσαλονίκης:** ΠΑΟΚ, Άρης, Ηρακλής.
- Οι υπόλοιπες οκτώ ομάδες δεν εντάσσονται σε κάποια από τις δύο κατηγορίες.
- Η κατηγοριοποίηση αυτή χρησιμοποιείται για τον προσδιορισμό των αναμετρήσεων υψηλής επικινδυνότητας και την εφαρμογή πρόσθετων περιορισμών ασφαλείας κατά τον χρονικό προγραμματισμό.

**Μαθηματική Διατύπωση**

Ορίζουμε τα ακόλουθα υποσύνολα του συνόλου ομάδων $T$:

$$
T_A=\{\text{Παναθηναϊκός, Ολυμπιακός, ΑΕΚ}\}
$$

$$
T_\Theta=\{\text{ΠΑΟΚ, Άρης, Ηρακλής}\}
$$

όπου $T_A$ και $T_\Theta$ αντιπροσωπεύουν τις ομάδες υψηλού κινδύνου Αθήνας και Θεσσαλονίκης, αντίστοιχα.

Τα δύο σύνολα είναι ξένα μεταξύ τους:

$$
T_A\cap T_\Theta=\varnothing
$$

Το σύνολο των υπόλοιπων ομάδων ορίζεται ως:

$$
T_O=T\setminus(T_A\cup T_\Theta)
$$

με:

$$
|T_A|=3,\qquad |T_\Theta|=3,\qquad |T_O|=8
$$




**Δυναμικότητα Ομάδων**

Για την εκτίμηση της αγωνιστικής ισχύος κάθε ομάδας εισάγεται ένας αριθμητικός δείκτης δυναμικότητας, ο οποίος συνδυάζει δύο βασικούς παράγοντες:

- **Την ιστορική αγωνιστική επίδοση**, όπως αποτυπώνεται από την τελική κατάταξη της ομάδας στα τρία προηγούμενα πρωταθλήματα.
- **Την τρέχουσα χρηματιστηριακή αξία του ρόστερ**, η οποία χρησιμοποιείται ως προσεγγιστικός δείκτης της σημερινής αγωνιστικής και οικονομικής ισχύος.

Η κατασκευή του δείκτη πραγματοποιείται σταδιακά, ξεκινώντας από την αξιολόγηση της ιστορικής επίδοσης και ενσωματώνοντας στη συνέχεια την οικονομική διάσταση.

**1. Ιστορική Αγωνιστική Επίδοση**

- Η ιστορική επίδοση κάθε ομάδας αξιολογείται με βάση την τελική της κατάταξη στις τρεις προηγούμενες αγωνιστικές περιόδους (2023–24, 2024–25 και 2025–26).
- Οι θέσεις κατάταξης μετατρέπονται σε αριθμητική βαθμολογία, ώστε οι υψηλότερες θέσεις να αντιστοιχούν σε μεγαλύτερη δυναμικότητα.
- Οι πιο πρόσφατες αγωνιστικές περίοδοι λαμβάνουν μεγαλύτερη βαρύτητα, καθώς θεωρούνται περισσότερο αντιπροσωπευτικές της σημερινής αγωνιστικής ισχύος.
- Σε περίπτωση που μια ομάδα δεν συμμετείχε στην πρώτη κατηγορία κατά κάποια από τις εξεταζόμενες περιόδους, της αποδίδεται συμβολικά η 15η θέση, δηλαδή μία θέση χαμηλότερα από την τελευταία του πρωταθλήματος.

**Μαθηματική Διατύπωση**

Έστω $p_{i,y}$ η τελική θέση της ομάδας $i$ στην αγωνιστική περίοδο $y$.

Ορίζουμε τη συνάρτηση βαθμολόγησης της ιστορικής κατάταξης:

$$
P(p)=100\frac{14-p}{13}
$$

Η συνάρτηση αυτή αντιστοιχίζει την πρώτη θέση σε 100 μονάδες και τη 14η θέση σε 0 μονάδες:

$$
P(1)=100,\qquad P(14)=0
$$

Κάθε βελτίωση μίας θέσης αυξάνει τη βαθμολογία κατά:

$$
\Delta P=\frac{100}{13}\approx7.69
$$

Για τις ομάδες που δεν συμμετείχαν στην πρώτη κατηγορία κατά κάποια αγωνιστική περίοδο, ορίζεται:

$$
p_{i,y}=15
$$

Επομένως:

$$
P(15)=-\frac{100}{13}\approx-7.69
$$

Η αρνητική τιμή εκφράζει την παραδοχή ότι η απουσία από την πρώτη κατηγορία αξιολογείται χαμηλότερα από την κατάληψη της τελευταίας θέσης.

Στη συνέχεια, υπολογίζεται η σταθμισμένη ιστορική επίδοση της ομάδας $i$:

$$
H_i=
0.20P(p_{i,23/24})
+0.30P(p_{i,24/25})
+0.50P(p_{i,25/26})
$$

Οι συντελεστές $0.20$, $0.30$ και $0.50$ αποδίδουν αυξανόμενη βαρύτητα στις πιο πρόσφατες αγωνιστικές περιόδους, με την τελευταία να συνεισφέρει κατά 50% στη συνολική ιστορική αξιολόγηση.

**2. Ενσωμάτωση της Χρηματιστηριακής Αξίας**

Η ιστορική κατάταξη δεν επαρκεί από μόνη της για την αποτύπωση της σημερινής δυναμικότητας μιας ομάδας. Μεταβολές στη σύνθεση του ρόστερ, μεταγραφικές επενδύσεις και αλλαγές στην οικονομική κατάσταση ενός συλλόγου μπορούν να διαφοροποιήσουν σημαντικά την αγωνιστική του ισχύ σε σχέση με τις προηγούμενες περιόδους.

Για τον λόγο αυτό, στον δείκτη δυναμικότητας ενσωματώνεται η συνολική χρηματιστηριακή αξία του ρόστερ, όπως εκτιμάται από την πλατφόρμα Transfermarkt. Η αξία αυτή χρησιμοποιείται ως προσεγγιστικό μέτρο της τρέχουσας ισχύος μιας ομάδας και δεν ταυτίζεται με τον πραγματικό μεταγραφικό ή λειτουργικό προϋπολογισμό της.

Επιλέγεται λογαριθμικός μετασχηματισμός της χρηματιστηριακής αξίας, ώστε η συνεισφορά της να αυξάνεται με φθίνοντα ρυθμό. Με τον τρόπο αυτό, περιορίζεται η επίδραση πολύ μεγάλων οικονομικών διαφορών μεταξύ των ομάδων, χωρίς να εξαλείφεται το πλεονέκτημα των συλλόγων με υψηλότερη αποτίμηση.

**Μαθηματική Διατύπωση**

Έστω $B_i>0$ η συνολική χρηματιστηριακή αξία του ρόστερ της ομάδας $i$, εκφρασμένη σε εκατομμύρια ευρώ.

Ορίζουμε την οικονομική συνιστώσα της δυναμικότητας:

$$
E_i=7.5\log_2(B_i)
$$

Η χρήση λογαρίθμου με βάση το $2$ σημαίνει ότι κάθε διπλασιασμός της χρηματιστηριακής αξίας αυξάνει την οικονομική συνιστώσα κατά $7.5$ μονάδες:

$$
E_i(2B_i)-E_i(B_i)=7.5
$$

**3. Τελικός Δείκτης Δυναμικότητας**

Ο τελικός δείκτης δυναμικότητας προκύπτει από τον συνδυασμό της σταθμισμένης ιστορικής επίδοσης και της λογαριθμικής χρηματιστηριακής αξίας.

Η ιστορική συνιστώσα πολλαπλασιάζεται με τον συντελεστή $0.30$, ενώ η οικονομική συνιστώσα διατηρεί τον συντελεστή $7.5$. Οι συντελεστές αυτοί καθορίζουν τη σχετική επιρροή των δύο παραγόντων στον τελικό δείκτη και αποτελούν παραδοχές της παρούσας εργασίας.

**Τελική Μαθηματική Διατύπωση**

Για κάθε ομάδα $i\in T$, ο δείκτης δυναμικότητας ορίζεται ως:

$$
\boxed{
F_i=0.30H_i+7.5\log_2(B_i)
}
$$

όπου:

$$
H_i=
0.20P(p_{i,23/24})
+0.30P(p_{i,24/25})
+0.50P(p_{i,25/26})
$$

και:

$$
P(p)=100\frac{14-p}{13}
$$

Για την καλύτερη κατανόηση της επίδρασης των συντελεστών:

- Κάθε διπλασιασμός της χρηματιστηριακής αξίας του ρόστερ αυξάνει τον τελικό δείκτη κατά **7.5 μονάδες**.
- Μία θέση καλύτερη στην κατάταξη της περιόδου 2025–26, χωρίς μεταβολή των άλλων παραγόντων, αυξάνει τον τελικό δείκτη κατά περίπου **1.15 μονάδες**.
- Μία θέση καλύτερη και στις τρεις εξεταζόμενες περιόδους αυξάνει τον τελικό δείκτη κατά περίπου **2.31 μονάδες**.

Ο δείκτης $F_i$ αποτελεί μια προσεγγιστική αποτύπωση της δυναμικότητας κάθε ομάδας, βασισμένη σε ιστορικά αγωνιστικά δεδομένα και στην τρέχουσα αποτίμηση του ρόστερ της. Οι συντελεστές του αποτελούν επιλογές μοντελοποίησης και όχι αποτελέσματα στατιστικής εκτίμησης.

Η δυναμικότητα αυτή χρησιμοποιείται στη συνέχεια για τον προσδιορισμό της εκτιμώμενης τηλεοπτικής ελκυστικότητας των αναμετρήσεων αλλά και την αξία των τηλεοπτικών παραθύρων.


### 2.3. Χρονικές Ζώνες και Τηλεοπτική Αξία

**Διαθέσιμες Χρονικές Ζώνες**

- Κάθε αναμέτρηση του πρωταθλήματος πρέπει να προγραμματιστεί σε μία από τις οκτώ διαθέσιμες τηλεοπτικές χρονικές ζώνες (*time slots*).
- Οι χρονικές ζώνες κατανέμονται σε τρεις ημέρες, Σάββατο, Κυριακή και Δευτέρα, και αντιπροσωπεύουν επιτρεπόμενα χρονικά παράθυρα έναρξης των αναμετρήσεων.
- Κάθε χρονική ζώνη μπορεί να φιλοξενήσει το πολύ μία αναμέτρηση ανά αγωνιστική. Επομένως, από τις οκτώ διαθέσιμες ζώνες χρησιμοποιούνται ακριβώς επτά σε κάθε αγωνιστική, ενώ μία παραμένει κενή.
- Σε κάθε χρονική ζώνη αποδίδεται μια προκαθορισμένη αριθμητική αξία, η οποία εκφράζει την εκτιμώμενη τηλεοπτική ελκυστικότητα της συγκεκριμένης ημέρας και ώρας.
- Οι τηλεοπτικές αξίες αποτελούν παραδοχές του μοντέλου και δεν προκύπτουν από πραγματικές μετρήσεις τηλεθέασης.

**Μαθηματική Διατύπωση**

Ορίζουμε το σύνολο των διαθέσιμων χρονικών ζωνών:

$$
S=\{S_1,S_2,\ldots,S_8\}
$$

Για κάθε χρονική ζώνη $s\in S$, ορίζεται η παράμετρος:

$$
V_s\in[0,1]
$$

όπου $V_s$ εκφράζει την εκτιμώμενη τηλεοπτική αξία της χρονικής ζώνης $s$.

Οι διαθέσιμες χρονικές ζώνες και οι αντίστοιχες αξίες τους παρουσιάζονται στον ακόλουθο πίνακα:

| Slot | Ημέρα | Παράθυρο έναρξης | $V_s$ |
|---|---|---|---|
| $S_1$ | Σάββατο | 16:00–17:00 | 0.40 |
| $S_2$ | Σάββατο | 18:00–19:00 | 0.50 |
| $S_3$ | Σάββατο | 20:00–21:00 | 0.80 |
| $S_4$ | Κυριακή | 16:00–17:00 | 0.70 |
| $S_5$ | Κυριακή | 17:00–18:00 | 0.60 |
| $S_6$ | Κυριακή | 18:00–19:00 | 0.90 |
| $S_7$ | Κυριακή | 20:00–21:00 | 1.00 |
| $S_8$ | Δευτέρα | 18:00–19:00 | 0.30 |

Για τη μαθηματική αναπαράσταση της χρονικής ανάθεσης, ορίζουμε τη δυαδική μεταβλητή απόφασης:

$$
y_{ijrs}=
\begin{cases}
1, & \text{αν η ομάδα }i\text{ υποδέχεται την }j\\
   & \text{στην αγωνιστική }r\text{ και στο slot }s\\
0, & \text{διαφορετικά}
\end{cases}
$$

για $i,j\in T$, $i\neq j$, $r\in R$ και $s\in S$.

Κάθε προγραμματισμένη αναμέτρηση ανατίθεται σε ακριβώς μία χρονική ζώνη:

$$
\sum_{s\in S}y_{ijrs}=x_{ijr}
\qquad
\forall i,j\in T,\ i\neq j,\ \forall r\in R
$$

Επιπλέον, κάθε χρονική ζώνη μπορεί να φιλοξενήσει το πολύ έναν αγώνα ανά αγωνιστική:

$$
\sum_{\substack{i,j\in T\\i\neq j}}y_{ijrs}\leq1
\qquad
\forall r\in R,\ \forall s\in S
$$

Δεδομένου ότι κάθε αγωνιστική περιλαμβάνει ακριβώς επτά αναμετρήσεις, οι παραπάνω περιορισμοί εξασφαλίζουν ότι επτά από τις οκτώ χρονικές ζώνες χρησιμοποιούνται και μία παραμένει διαθέσιμη.

Οι παράμετροι $V_s$ αξιοποιούνται στη συνέχεια για τον υπολογισμό της εκτιμώμενης τηλεοπτικής αξίας των αναμετρήσεων και τη διαμόρφωση της αντικειμενικής συνάρτησης του προβλήματος.


### 2.4. Τηλεοπτική Ελκυστικότητα Αναμετρήσεων

Η εκτιμώμενη τηλεοπτική ελκυστικότητα μιας αναμέτρησης αποτελεί βασική παράμετρο του προβλήματος βελτιστοποίησης, καθώς επιτρέπει την ποσοτική αξιολόγηση των αγώνων με βάση τις συμμετέχουσες ομάδες και τη χρονική ζώνη διεξαγωγής τους.

Για τον προσδιορισμό της αξιοποιείται ο δείκτης δυναμικότητας $F_i$, ο οποίος ορίστηκε στην ενότητα 2.2 και συνδυάζει την ιστορική αγωνιστική επίδοση με τη χρηματιστηριακή αξία των ομάδων.

Η διαδικασία περιλαμβάνει τρία στάδια:

- **Βασική τηλεοπτική ελκυστικότητα ομάδων:** Η δυναμικότητα κάθε ομάδας μετατρέπεται σε έναν κανονικοποιημένο δείκτη ελκυστικότητας.
- **Διαφοροποίηση ανά αγωνιστική:** Εισάγεται περιορισμένη τυχαία διακύμανση γύρω από τη βασική ελκυστικότητα κάθε ομάδας.
- **Ελκυστικότητα αναμέτρησης:** Η ελκυστικότητα κάθε αγώνα προκύπτει από το άθροισμα των επιμέρους τιμών των δύο συμμετεχουσών ομάδων.

**1. Βασική Τηλεοπτική Ελκυστικότητα Ομάδων**

- Η αγωνιστική δυναμικότητα χρησιμοποιείται ως προσεγγιστικός δείκτης της τηλεοπτικής ελκυστικότητας μιας ομάδας, με την παραδοχή ότι ομάδες υψηλότερης αγωνιστικής και οικονομικής ισχύος τείνουν να συγκεντρώνουν μεγαλύτερο ενδιαφέρον.
- Οι τιμές δυναμικότητας μετατρέπονται σε κοινή αριθμητική κλίμακα μέσω γραμμικής κανονικοποίησης.
- Η βασική τηλεοπτική ελκυστικότητα περιορίζεται στο διάστημα $[0.2,0.9]$, εξασφαλίζοντας θετική αξία ακόμη και για τις ομάδες χαμηλότερης δυναμικότητας.

**Μαθηματική Διατύπωση**

Έστω $F_i$ η δυναμικότητα της ομάδας $i\in T$.

Ορίζουμε τις ακραίες τιμές δυναμικότητας:

$$
F_{\min}=\min_{i\in T}F_i
$$

$$
F_{\max}=\max_{i\in T}F_i
$$

Η βασική τηλεοπτική ελκυστικότητα $\mu_i$ της ομάδας $i$ υπολογίζεται ως:

$$
\boxed{
\mu_i=0.2+0.7\frac{F_i-F_{\min}}{F_{\max}-F_{\min}}
}
$$

για $F_{\max}>F_{\min}$.

Στην ειδική περίπτωση όπου όλες οι ομάδες έχουν την ίδια δυναμικότητα, ορίζεται:

$$
\mu_i=0.55,\qquad \forall i\in T
$$

Η κανονικοποίηση εξασφαλίζει ότι:

$$
0.2\leq\mu_i\leq0.9
$$

Η ομάδα με τη μικρότερη δυναμικότητα λαμβάνει βασική ελκυστικότητα $0.2$, ενώ η ομάδα με τη μεγαλύτερη δυναμικότητα λαμβάνει $0.9$. Οι υπόλοιπες τιμές κατανέμονται γραμμικά μεταξύ των δύο ορίων.

Το συγκεκριμένο εύρος αποτελεί παραδοχή του μοντέλου και επιτρέπει τη διαφοροποίηση των ομάδων χωρίς να αποδίδεται μηδενική ελκυστικότητα στις ασθενέστερες.

**2. Διαφοροποίηση της Τηλεοπτικής Ελκυστικότητας ανά Αγωνιστική**

- Η τηλεοπτική ελκυστικότητα μιας ομάδας δεν θεωρείται απολύτως σταθερή σε ολόκληρη τη διάρκεια του πρωταθλήματος.
- Για την αναπαράσταση περιορισμένης μεταβλητότητας εισάγεται τυχαία διακύμανση γύρω από τη βασική ελκυστικότητα κάθε ομάδας.
- Η διακύμανση ακολουθεί ομοιόμορφη κατανομή και επιτρέπει διαφορετική τιμή ελκυστικότητας για κάθε ομάδα και αγωνιστική.
- Οι τιμές παράγονται μία φορά πριν από τη βελτιστοποίηση και παραμένουν σταθερές καθ' όλη τη διάρκεια της επίλυσης.

**Μαθηματική Διατύπωση**

Έστω $a_{i,r}$ η εκτιμώμενη τηλεοπτική ελκυστικότητα της ομάδας $i$ κατά την αγωνιστική $r$.

Ορίζουμε:

$$
\boxed{
a_{i,r}\sim\mathcal{U}(\mu_i-0.1,\mu_i+0.1)
}
$$

για κάθε $i\in T$ και $r\in R$.

Η $\mathcal{U}(a,b)$ συμβολίζει τη συνεχή ομοιόμορφη κατανομή στο διάστημα $[a,b]$.

Επομένως:

$$
\mu_i-0.1\leq a_{i,r}\leq\mu_i+0.1
$$

Δεδομένου ότι $\mu_i\in[0.2,0.9]$, προκύπτει:

$$
\boxed{0.1\leq a_{i,r}\leq1}
$$

Η αναμενόμενη τιμή της ελκυστικότητας ισούται με τη βασική ελκυστικότητα της ομάδας:

$$
\mathbb{E}[a_{i,r}]=\mu_i
$$

Για τις 14 ομάδες και τις 26 αγωνιστικές παράγονται συνολικά:

$$
|T|\cdot|R|=14\cdot26=364
$$

τιμές τηλεοπτικής ελκυστικότητας.

Οι τιμές αυτές καθορίζονται πριν ξεκινήσει η διαδικασία βελτιστοποίησης και αποτελούν σταθερές παραμέτρους του προβλήματος.

Συνεπώς, παρά την τυχαία διαδικασία παραγωγής των τιμών, η βελτιστοποίηση πραγματοποιείται πάνω σε ένα προκαθορισμένο σενάριο και αντιμετωπίζεται ως **ντετερμινιστικό πρόβλημα συνδυαστικής βελτιστοποίησης**.

**3. Τηλεοπτική Ελκυστικότητα Αναμέτρησης**

- Η τηλεοπτική ελκυστικότητα κάθε αναμέτρησης εξαρτάται από την ελκυστικότητα των δύο συμμετεχουσών ομάδων.
- Επιλέγεται προσθετική σχέση, σύμφωνα με την οποία κάθε ομάδα συνεισφέρει ανεξάρτητα στην εκτιμώμενη ελκυστικότητα του αγώνα.
- Η σχέση είναι συμμετρική ως προς τις δύο ομάδες, επομένως η επιλογή γηπεδούχου δεν επηρεάζει άμεσα την ελκυστικότητα της αναμέτρησης.
- Δεν εισάγεται πρόσθετος συντελεστής ελκυστικότητας για ντέρμπι ή συγκεκριμένους συνδυασμούς ομάδων.

**Μαθηματική Διατύπωση**

Έστω $A_{ijr}$ η εκτιμώμενη τηλεοπτική ελκυστικότητα της αναμέτρησης μεταξύ των ομάδων $i$ και $j$, κατά την αγωνιστική $r$.

Ορίζουμε:

$$
\boxed{
A_{ijr}=a_{i,r}+a_{j,r}
}
$$

για κάθε $i,j\in T$, με $i\neq j$, και $r\in R$.

Από τα όρια της ελκυστικότητας των ομάδων προκύπτει:

$$
\boxed{
0.2\leq A_{ijr}\leq2
}
$$

Επιπλέον, ισχύει η συμμετρική ιδιότητα:

$$
A_{ijr}=A_{jir}
$$

Η τελική εκτιμώμενη τηλεοπτική αξία μιας αναμέτρησης εξαρτάται επιπλέον από τη χρονική ζώνη στην οποία προγραμματίζεται.

Χρησιμοποιώντας την προκαθορισμένη αξία $V_s$ της χρονικής ζώνης $s$, όπως ορίστηκε στην ενότητα 2.3, υπολογίζουμε:

$$
\boxed{
W_{ijrs}=A_{ijr}V_s
}
$$

Αντικαθιστώντας την ελκυστικότητα της αναμέτρησης:

$$
\boxed{
W_{ijrs}=(a_{i,r}+a_{j,r})V_s
}
$$

για κάθε $i,j\in T$, $i\neq j$, $r\in R$ και $s\in S$.

Με τον τρόπο αυτό, η ίδια αναμέτρηση αποκτά διαφορετική εκτιμώμενη τηλεοπτική αξία ανάλογα με τη χρονική ζώνη στην οποία προγραμματίζεται.

Η τελική αξία κάθε αγώνα καθορίζεται επομένως από δύο παράγοντες:

- Την εκτιμώμενη ελκυστικότητα των συμμετεχουσών ομάδων κατά τη συγκεκριμένη αγωνιστική.
- Την προκαθορισμένη τηλεοπτική αξία της χρονικής ζώνης διεξαγωγής.

Οι ποσότητες $W_{ijrs}$ χρησιμοποιούνται στη συνέχεια για τον υπολογισμό της συνολικής τηλεοπτικής αξίας κάθε αγωνιστικής και τη διαμόρφωση της αντικειμενικής συνάρτησης του προβλήματος.

**Μεθοδολογική Παρατήρηση**

Οι δείκτες τηλεοπτικής ελκυστικότητας αποτελούν προσεγγίσεις που κατασκευάζονται στο πλαίσιο του μοντέλου και δεν προκύπτουν από πραγματικές μετρήσεις τηλεθέασης.

Ειδικότερα, η αγωνιστική δυναμικότητα χρησιμοποιείται ως υποκατάστατος δείκτης της δημοτικότητας των ομάδων, ενώ η τυχαία διακύμανση ανά αγωνιστική αποτελεί παραδοχή μοντελοποίησης και δεν αναπαριστά παρατηρημένες μεταβολές της φόρμας ή του ενδιαφέροντος των τηλεθεατών.

Κατά συνέπεια, η εκτιμώμενη τηλεοπτική αξία χρησιμοποιείται για τη σχετική αξιολόγηση εναλλακτικών προγραμμάτων, χωρίς να αποτελεί πρόβλεψη πραγματικής τηλεθέασης ή τηλεοπτικών εσόδων.




### 2.5. Πρόσθετοι Περιορισμοί Προγραμματισμού

Πέρα από τους βασικούς περιορισμούς που καθορίζουν τη δομή του πρωταθλήματος και την ανάθεση των αναμετρήσεων στις διαθέσιμες χρονικές ζώνες, το μοντέλο ενσωματώνει πρόσθετους περιορισμούς που αποσκοπούν στη διαμόρφωση ενός οργανωτικά εφικτού και ισορροπημένου αγωνιστικού προγράμματος.

Οι περιορισμοί αυτοί αφορούν τη γεωγραφική κατανομή των εντός έδρας αναμετρήσεων, την αποφυγή μεγάλων ακολουθιών εντός ή εκτός έδρας αγώνων και την αποτροπή άμεσης επανάληψης μιας αναμέτρησης κατά τη μετάβαση μεταξύ των δύο γύρων.

Οι περιορισμοί αποτελούν παραδοχές της παρούσας εφαρμογής και δεν παρουσιάζονται ως υποχρεωτικοί επίσημοι κανονισμοί της Super League.

**1. Γεωγραφικοί Περιορισμοί και Ασφάλεια**

- Για τον περιορισμό πιθανών ζητημάτων ασφάλειας, δεν επιτρέπεται ο ταυτόχρονος προγραμματισμός περισσότερων από μίας εντός έδρας αναμετρήσεων ομάδων υψηλού κινδύνου της ίδιας γεωγραφικής κατηγορίας, κατά την ίδια ημέρα και αγωνιστική.
- Η κατηγοριοποίηση των ομάδων υψηλού κινδύνου σε Αθήνα και Θεσσαλονίκη έχει ήδη οριστεί στην ενότητα 2.2.
- Ο περιορισμός εφαρμόζεται ξεχωριστά στις δύο γεωγραφικές κατηγορίες και δεν αποκλείει τη διεξαγωγή εντός έδρας αγώνων ομάδων διαφορετικών κατηγοριών την ίδια ημέρα.

**Μαθηματική Διατύπωση**

Υπενθυμίζουμε τα σύνολα ομάδων υψηλού κινδύνου:

$$
T_A=\{\text{Παναθηναϊκός, Ολυμπιακός, ΑΕΚ}\}
$$

$$
T_\Theta=\{\text{ΠΑΟΚ, Άρης, Ηρακλής}\}
$$

Ορίζουμε το σύνολο των ημερών διεξαγωγής:

$$
D=\{\text{Σάββατο, Κυριακή, Δευτέρα}\}
$$

Για κάθε ημέρα $d\in D$, έστω $S_d\subseteq S$ το σύνολο των χρονικών ζωνών που αντιστοιχούν στη συγκεκριμένη ημέρα.

Με χρήση της δυαδικής μεταβλητής $y_{ijrs}$, που ορίστηκε στην ενότητα 2.3, διατυπώνουμε:

$$
\boxed{
\sum_{i\in T_A}
\sum_{\substack{j\in T\\j\neq i}}
\sum_{s\in S_d}
y_{ijrs}
\leq 1
}
$$

$$
\forall r\in R,\quad \forall d\in D
$$

Αντίστοιχα, για τις ομάδες υψηλού κινδύνου της Θεσσαλονίκης:

$$
\boxed{
\sum_{i\in T_\Theta}
\sum_{\substack{j\in T\\j\neq i}}
\sum_{s\in S_d}
y_{ijrs}
\leq 1
}
$$

$$
\forall r\in R,\quad \forall d\in D
$$

Οι παραπάνω περιορισμοί εξασφαλίζουν ότι, σε κάθε αγωνιστική και ημέρα, το πολύ μία ομάδα από κάθε γεωγραφική κατηγορία υψηλού κινδύνου μπορεί να αγωνίζεται ως γηπεδούχος.

**2. Περιορισμός Συνεχόμενων Εντός και Εκτός Έδρας Αγώνων**

- Κάθε ομάδα επιτρέπεται να πραγματοποιήσει το πολύ δύο συνεχόμενες αναμετρήσεις εντός έδρας ή δύο συνεχόμενες εκτός έδρας.
- Ο περιορισμός εφαρμόζεται σε ολόκληρη τη διάρκεια του πρωταθλήματος, συμπεριλαμβανομένης της μετάβασης από τον πρώτο στον δεύτερο γύρο.
- Με τον τρόπο αυτό αποφεύγονται μεγάλες συνεχόμενες ακολουθίες εντός ή εκτός έδρας αγώνων.

**Μαθηματική Διατύπωση**

Για κάθε ομάδα $i\in T$ και αγωνιστική $r\in R$, ορίζουμε τη δυαδική ένδειξη εντός έδρας αγώνα:

$$
h_{ir}=
\sum_{\substack{j\in T\\j\neq i}}x_{ijr}
$$

Από τον περιορισμό συμμετοχής κάθε ομάδας σε ακριβώς έναν αγώνα ανά αγωνιστική προκύπτει:

$$
h_{ir}=
\begin{cases}
1, & \text{αν η ομάδα }i\text{ αγωνίζεται εντός έδρας}\\
0, & \text{αν η ομάδα }i\text{ αγωνίζεται εκτός έδρας}
\end{cases}
$$

Για κάθε τριάδα διαδοχικών αγωνιστικών επιβάλλεται:

$$
\boxed{
1\leq
h_{ir}+h_{i,r+1}+h_{i,r+2}
\leq2
}
$$

$$
\forall i\in T,\quad r=1,\ldots,24
$$

Το κάτω όριο εξασφαλίζει ότι καμία ομάδα δεν πραγματοποιεί τρεις συνεχόμενες εκτός έδρας αναμετρήσεις, ενώ το άνω όριο αποκλείει τρεις συνεχόμενες εντός έδρας αναμετρήσεις.

Ο περιορισμός εφαρμόζεται και σε τριάδες αγωνιστικών που εκτείνονται μεταξύ των δύο γύρων, όπως οι αγωνιστικές 12–14 και 13–15.

**3. Περιορισμός Άμεσης Επανάληψης Αναμέτρησης μεταξύ των Δύο Γύρων**

- Κάθε ζευγάρι ομάδων αναμετράται μία φορά σε κάθε γύρο, με αντίστροφες έδρες, όπως καθορίστηκε στην ενότητα 2.1.
- Παρότι η σειρά των αναμετρήσεων στον δεύτερο γύρο δεν είναι υποχρεωτικά ίδια με εκείνη του πρώτου, δεν επιτρέπεται το ίδιο ζευγάρι ομάδων να αγωνιστεί στην τελευταία αγωνιστική του πρώτου γύρου και αμέσως ξανά στην πρώτη αγωνιστική του δεύτερου.
- Ο περιορισμός αποσκοπεί στην αποφυγή δύο διαδοχικών συναντήσεων των ίδιων ομάδων κατά τη μετάβαση μεταξύ των γύρων.

**Μαθηματική Διατύπωση**

Για κάθε μη διατεταγμένο ζευγάρι ομάδων $i,j\in T$, με $i<j$, επιβάλλεται:

$$
\boxed{
x_{ij,13}+x_{ji,13}
+x_{ij,14}+x_{ji,14}
\leq1
}
$$

$$
\forall i,j\in T,\quad i<j
$$

Ο περιορισμός εξασφαλίζει ότι, εάν δύο ομάδες αναμετρηθούν στη 13η αγωνιστική, δεν μπορούν να αναμετρηθούν ξανά στη 14η, ανεξάρτητα από την αντιστροφή των εδρών.

Δεν επιβάλλεται πρόσθετος περιορισμός ελάχιστης απόστασης για αναμετρήσεις που διεξάγονται σε μη διαδοχικές αγωνιστικές μεταξύ των δύο γύρων.

**Συνολική Ερμηνεία των Περιορισμών**

Οι παραπάνω περιορισμοί λειτουργούν συμπληρωματικά προς τους βασικούς κανόνες της διοργάνωσης και της χρονικής ανάθεσης που παρουσιάστηκαν στις ενότητες 2.1 και 2.3.

Συγκεκριμένα, το μοντέλο απαιτεί:

- Την τήρηση γεωγραφικών περιορισμών για τις εντός έδρας αναμετρήσεις ομάδων υψηλού κινδύνου.
- Την αποφυγή τριών συνεχόμενων εντός ή εκτός έδρας αναμετρήσεων για οποιαδήποτε ομάδα.
- Την αποτροπή άμεσης επανάληψης της ίδιας αναμέτρησης στη μετάβαση από τη 13η στη 14η αγωνιστική.

Όλοι οι παραπάνω περιορισμοί αντιμετωπίζονται ως υποχρεωτικοί (*hard constraints*). Επομένως, κάθε αποδεκτή λύση πρέπει να τους ικανοποιεί πλήρως, ανεξάρτητα από την τηλεοπτική αξία ή την αντικειμενική τιμή του προγράμματος.

Η βελτιστοποίηση πραγματοποιείται αποκλειστικά μεταξύ των προγραμμάτων που ικανοποιούν το σύνολο των περιορισμών εφικτότητας.



### 2.6. Αντικειμενική Συνάρτηση

Η αντικειμενική συνάρτηση του προβλήματος επιδιώκει την κατασκευή ενός αγωνιστικού προγράμματος που συνδυάζει υψηλή συνολική τηλεοπτική αξία με ισορροπημένη κατανομή της ελκυστικότητας μεταξύ των αγωνιστικών.

Η μεγιστοποίηση αποκλειστικά της συνολικής τηλεοπτικής αξίας θα μπορούσε να οδηγήσει σε προγράμματα όπου ορισμένες αγωνιστικές συγκεντρώνουν ιδιαίτερα ελκυστικές αναμετρήσεις, ενώ άλλες παρουσιάζουν σημαντικά χαμηλότερη τηλεοπτική αξία.

Για τον λόγο αυτό, η αντικειμενική συνάρτηση περιλαμβάνει δύο συνιστώσες:

- **Μεγιστοποίηση συνολικής τηλεοπτικής αξίας:** Επιδιώκεται η τοποθέτηση των αναμετρήσεων σε χρονικές ζώνες που μεγιστοποιούν τη συνολική εκτιμώμενη τηλεοπτική αξία του πρωταθλήματος.
- **Εξισορρόπηση τηλεοπτικής αξίας μεταξύ αγωνιστικών:** Επιβάλλεται ποινή στη διαφορά μεταξύ της αγωνιστικής με τη μεγαλύτερη και εκείνης με τη μικρότερη εκτιμώμενη τηλεοπτική αξία.

Η σχετική σημασία των δύο στόχων καθορίζεται από έναν μη αρνητικό συντελεστή βαρύτητας $\lambda$.

**1. Υπολογισμός Τηλεοπτικής Αξίας ανά Αγωνιστική**

Η εκτιμώμενη τηλεοπτική αξία κάθε αναμέτρησης έχει οριστεί στην ενότητα 2.4 ως:

$$
W_{ijrs}=(a_{i,r}+a_{j,r})V_s
$$

Η συνολική τηλεοπτική αξία μιας αγωνιστικής προκύπτει από το άθροισμα των αξιών των επτά αναμετρήσεων που έχουν προγραμματιστεί σε αυτή.

**Μαθηματική Διατύπωση**

Έστω $U_r$ η συνολική εκτιμώμενη τηλεοπτική αξία της αγωνιστικής $r\in R$.

Χρησιμοποιώντας τη δυαδική μεταβλητή ανάθεσης $y_{ijrs}$, ορίζουμε:

$$
\boxed{
U_r=
\sum_{\substack{i,j\in T\\i\neq j}}
\sum_{s\in S}
(a_{i,r}+a_{j,r})V_s y_{ijrs}
}
$$

$$
\forall r\in R
$$

Η μεταβλητή $y_{ijrs}$ λαμβάνει την τιμή $1$ μόνο όταν η ομάδα $i$ υποδέχεται την ομάδα $j$ στην αγωνιστική $r$ και στη χρονική ζώνη $s$.

Επομένως, στο άθροισμα συνεισφέρουν αποκλειστικά οι αναμετρήσεις που έχουν προγραμματιστεί στη συγκεκριμένη αγωνιστική.

Η συνολική εκτιμώμενη τηλεοπτική αξία του πρωταθλήματος είναι:

$$
\boxed{
U_{\mathrm{total}}=\sum_{r\in R}U_r
}
$$

**2. Μέτρηση Ανισορροπίας μεταξύ Αγωνιστικών**

Η συνολική τηλεοπτική αξία δεν αποτυπώνει τον τρόπο με τον οποίο κατανέμεται η ελκυστικότητα μεταξύ των 26 αγωνιστικών.

Για την αξιολόγηση της ισορροπίας εισάγεται ένας δείκτης που μετρά τη διαφορά μεταξύ της αγωνιστικής με τη μεγαλύτερη και εκείνης με τη μικρότερη τηλεοπτική αξία.

**Μαθηματική Διατύπωση**

Ορίζουμε:

$$
U_{\max}=\max_{r\in R}U_r
$$

$$
U_{\min}=\min_{r\in R}U_r
$$

Η ανισορροπία του προγράμματος εκφράζεται από τη διαφορά:

$$
\boxed{
\Delta U=U_{\max}-U_{\min}
}
$$

Ισχύει:

$$
\Delta U\geq0
$$

Μικρότερες τιμές του $\Delta U$ υποδηλώνουν πιο ισορροπημένη κατανομή της εκτιμώμενης τηλεοπτικής αξίας μεταξύ των αγωνιστικών.

Η επιλογή της διαφοράς μέγιστης και ελάχιστης αξίας επιτρέπει την αξιολόγηση της συνολικής διακύμανσης του προγράμματος ως προς τις ακραίες αγωνιστικές, χωρίς να απαιτείται η επιβολή συγκεκριμένου στόχου αξίας για κάθε αγωνιστική.

**3. Διαμόρφωση της Αντικειμενικής Συνάρτησης**

Η τελική αντικειμενική συνάρτηση συνδυάζει τη μεγιστοποίηση της συνολικής τηλεοπτικής αξίας με την ελαχιστοποίηση της ανισορροπίας μεταξύ αγωνιστικών.

Η δεύτερη συνιστώσα εισάγεται ως ποινή, η οποία αφαιρείται από τη συνολική αξία του προγράμματος.

**Μαθηματική Διατύπωση**

Ορίζουμε τον μη αρνητικό συντελεστή βαρύτητας:

$$
\lambda\geq0
$$

Η αντικειμενική συνάρτηση διατυπώνεται ως:

$$
\boxed{
\max Z_{\lambda}
=
\sum_{r\in R}U_r
-
\lambda(U_{\max}-U_{\min})
}
$$

ή ισοδύναμα:

$$
\boxed{
\max Z_{\lambda}
=
U_{\mathrm{total}}-\lambda\Delta U
}
$$

Η παράμετρος $\lambda$ καθορίζει τη σχετική σημασία της ισορροπίας μεταξύ αγωνιστικών.

Ειδικότερα:

- Για $\lambda=0$, η βελτιστοποίηση αποσκοπεί αποκλειστικά στη μεγιστοποίηση της συνολικής τηλεοπτικής αξίας, χωρίς να λαμβάνεται υπόψη η ανισορροπία μεταξύ αγωνιστικών.
- Για $\lambda>0$, η διαφορά μεταξύ μέγιστης και ελάχιστης τηλεοπτικής αξίας επιβαρύνει την αντικειμενική συνάρτηση.
- Όσο αυξάνεται η τιμή του $\lambda$, τόσο μεγαλύτερη σημασία αποδίδεται στη μείωση της ανισορροπίας σε σχέση με τη μεγιστοποίηση της συνολικής αξίας.

Η επιλογή του $\lambda$ αποτελεί παράμετρο του μοντέλου και επιτρέπει την εξέταση διαφορετικών συμβιβασμών μεταξύ των δύο στόχων.

**4. Γραμμική Αναπαράσταση της Ανισορροπίας**

Για την ενσωμάτωση της αντικειμενικής συνάρτησης σε μοντέλο ακέραιου προγραμματισμού, εισάγονται δύο βοηθητικές συνεχείς μεταβλητές:

- $u^{\max}$: άνω φράγμα της τηλεοπτικής αξίας όλων των αγωνιστικών.
- $u^{\min}$: κάτω φράγμα της τηλεοπτικής αξίας όλων των αγωνιστικών.

**Μαθηματική Διατύπωση**

Επιβάλλονται οι περιορισμοί:

$$
\boxed{
U_r\leq u^{\max}
\qquad \forall r\in R
}
$$

$$
\boxed{
U_r\geq u^{\min}
\qquad \forall r\in R
}
$$

Η αντικειμενική συνάρτηση γράφεται:

$$
\boxed{
\max Z_{\lambda}
=
\sum_{r\in R}U_r
-
\lambda(u^{\max}-u^{\min})
}
$$

Για $\lambda>0$, η μεγιστοποίηση της αντικειμενικής συνάρτησης ωθεί τις βοηθητικές μεταβλητές προς τις ακραίες πραγματικές τιμές των αγωνιστικών, ώστε:

$$
u^{\max}=\max_{r\in R}U_r
$$

$$
u^{\min}=\min_{r\in R}U_r
$$

Συνεπώς, η παραπάνω γραμμική αναπαράσταση είναι ισοδύναμη με την αρχική διατύπωση της αντικειμενικής συνάρτησης.

Στην υπολογιστική υλοποίηση με CP-SAT χρησιμοποιούνται κατάλληλα κλιμακωμένοι ακέραιοι συντελεστές, καθώς ο επιλυτής απαιτεί ακέραιες αριθμητικές εκφράσεις. Η διαδικασία αυτή προσεγγίζει την αρχική αντικειμενική συνάρτηση μέσω στρογγυλοποίησης των συντελεστών.

**5. Συνολική Ερμηνεία της Αντικειμενικής Συνάρτησης**

Η αντικειμενική συνάρτηση αξιολογεί κάθε εφικτό αγωνιστικό πρόγραμμα με βάση δύο αλληλοσυνδεόμενα κριτήρια:

- Τη συνολική εκτιμώμενη τηλεοπτική αξία που προκύπτει από την κατανομή των αναμετρήσεων στις διαθέσιμες χρονικές ζώνες.
- Την ισορροπία της τηλεοπτικής αξίας μεταξύ των αγωνιστικών, όπως αυτή μετράται από τη διαφορά μέγιστης και ελάχιστης αξίας.

Η ταυτόχρονη εξέταση των δύο κριτηρίων επιτρέπει την αναζήτηση προγραμμάτων που δεν συγκεντρώνουν απλώς υψηλή συνολική ελκυστικότητα, αλλά επιδιώκουν και την αποφυγή ιδιαίτερα μεγάλων διαφορών μεταξύ των αγωνιστικών.

Η βελτιστοποίηση πραγματοποιείται υπό το σύνολο των περιορισμών εφικτότητας που παρουσιάστηκαν στις προηγούμενες ενότητες.

Καθώς η τηλεοπτική ελκυστικότητα και οι αξίες των χρονικών ζωνών αποτελούν παραμέτρους του μοντέλου, τα αποτελέσματα εκφράζουν τη σχετική ποιότητα των προγραμμάτων υπό τις συγκεκριμένες παραδοχές και όχι προβλέψεις πραγματικής τηλεθέασης.



### 3.1. Συνολική Αρχιτεκτονική Επίλυσης

Για την επίλυση του προβλήματος προγραμματισμού της ελληνικής Super League αναπτύχθηκε μια υβριδική μέθοδος, η οποία συνδυάζει τον επιλυτή Constraint Programming – Satisfiability (CP-SAT) της βιβλιοθήκης Google OR-Tools με τον μεταευρετικό αλγόριθμο Adaptive Large Neighborhood Search (ALNS). Η προσέγγιση αποσκοπεί στην κατασκευή εφικτών αγωνιστικών προγραμμάτων και στη βελτίωση της αντικειμενικής τους αξίας, αξιοποιώντας τις δυνατότητες των δύο μεθόδων σε διαφορετικά στάδια της επίλυσης.

Αρχικά, ο CP-SAT αναλαμβάνει την κατασκευή ενός πλήρους αγωνιστικού προγράμματος που ικανοποιεί όλους τους υποχρεωτικούς περιορισμούς του μοντέλου. Η αναζήτηση τερματίζεται μόλις εντοπιστεί η πρώτη εφικτή λύση, η οποία χρησιμοποιείται ως αφετηρία της διαδικασίας βελτιστοποίησης. Στη συνέχεια, ο ALNS επιχειρεί να βελτιώσει το πρόγραμμα μέσω επαναλαμβανόμενων διαδικασιών μερικής αποδόμησης (destroy) και ανακατασκευής (repair). Η επιλογή των τμημάτων που αποδεσμεύονται πραγματοποιείται μέσω διαφορετικών destroy operators με προσαρμοστικά βάρη, ενώ η ανακατασκευή ανατίθεται εκ νέου στον CP-SAT. Η αξιολόγηση και αποδοχή των υποψήφιων λύσεων πραγματοποιείται με μηχανισμό Simulated Annealing, διατηρώντας παράλληλα την καλύτερη εφικτή λύση που έχει εντοπιστεί.

Τέλος, η αρχική και η τελική λύση υποβάλλονται σε ανεξάρτητο έλεγχο εφικτότητας και αξιολόγηση της αντικειμενικής τους αξίας. Η ποιότητα των αποτελεσμάτων εξετάζεται επιπλέον μέσω μαθηματικών άνω φραγμάτων, τα οποία επιτρέπουν την εκτίμηση της απόστασης από τη βέλτιστη δυνατή λύση για διαφορετικές τιμές της παραμέτρου ισορροπίας. Η αναλυτική περιγραφή των επιμέρους σταδίων ακολουθεί στις επόμενες ενότητες.



### 3.2. Κατασκευή Αρχικής Εφικτής Λύσης με CP-SAT

Το πρώτο στάδιο της διαδικασίας επίλυσης πραγματοποιείται μέσω του CP-SAT (Constraint Programming – Satisfiability), ενός επιλυτή διακριτής βελτιστοποίησης της βιβλιοθήκης Google OR-Tools. Ο CP-SAT είναι κατάλληλος για προβλήματα με μεγάλο πλήθος διακριτών αποφάσεων και αλληλοεξαρτώμενων περιορισμών, αξιοποιώντας τεχνικές διάδοσης περιορισμών, λογικών συμπερασμών και συστηματικής αναζήτησης.

Στην παρούσα υλοποίηση, ο CP-SAT χρησιμοποιεί το μαθηματικό μοντέλο του Κεφαλαίου 2 για να καθορίσει ταυτόχρονα τις αναμετρήσεις κάθε αγωνιστικής, τις έδρες και τις αντίστοιχες τηλεοπτικές χρονικές ζώνες. Κατά την κατασκευή του προγράμματος επιβάλλονται όλοι οι υποχρεωτικοί περιορισμοί, ώστε η παραγόμενη λύση να αποτελεί ένα πλήρες και εφικτό πρόγραμμα 182 αναμετρήσεων, κατανεμημένων σε 26 αγωνιστικές.

Παρότι το υπολογιστικό μοντέλο περιλαμβάνει την αντικειμενική συνάρτηση, στο συγκεκριμένο στάδιο η αναζήτηση διακόπτεται μόλις εντοπιστεί η πρώτη εφικτή λύση, χωρίς να απαιτείται περαιτέρω βελτιστοποίηση ή απόδειξη βελτιστότητας. Η διαδικασία υπόκειται σε προκαθορισμένο χρονικό όριο και αξιοποιεί τη δυνατότητα παράλληλης αναζήτησης του CP-SAT. Εάν δεν εντοπιστεί εφικτό πρόγραμμα εντός του διαθέσιμου χρόνου, η εκτέλεση δεν προχωρά στο επόμενο στάδιο.

Μετά την επιτυχή ολοκλήρωση της αρχικής αναζήτησης, το πρόγραμμα εξάγεται και υποβάλλεται σε ανεξάρτητο έλεγχο εφικτότητας. Εφόσον επιβεβαιωθεί η ορθότητά του, υπολογίζεται η αρχική αντικειμενική του τιμή και η λύση μεταφέρεται στον ALNS ως σημείο εκκίνησης της διαδικασίας βελτίωσης.



### 3.3. Βελτιστοποίηση μέσω Adaptive Large Neighborhood Search (ALNS)

Μετά την κατασκευή της αρχικής εφικτής λύσης, η διαδικασία βελτιστοποίησης συνεχίζεται με την εφαρμογή του Adaptive Large Neighborhood Search (ALNS). Ο αλγόριθμος εξερευνά τον χώρο των εφικτών προγραμμάτων μέσω επαναλαμβανόμενων διαδικασιών μερικής αποδόμησης (destroy) και ανακατασκευής (repair). Αντί να περιορίζεται σε μικρές τοπικές μεταβολές, επιτρέπει την ταυτόχρονη επανεξέταση μεγαλύτερων τμημάτων του αγωνιστικού προγράμματος, επιδιώκοντας τη βελτίωση της αντικειμενικής του αξίας.


**Destroy Operators.** Η υλοποίηση περιλαμβάνει τρεις διαφορετικούς μηχανισμούς αποδόμησης, οι οποίοι καθορίζουν τις αποφάσεις που αποδεσμεύονται σε κάθε επανάληψη:

- **Random Rounds:** Επιλέγει τυχαία έναν από τους δύο γύρους του πρωταθλήματος και αποδεσμεύει δύο έως τέσσερις τυχαία επιλεγμένες αγωνιστικές του, επιτρέποντας την ανακατανομή των αναμετρήσεων και των χρονικών ζωνών τους.

- **Worst-Value Rounds:** Επιλέγει τυχαία έναν από τους δύο γύρους και εντοπίζει τις πέντε αγωνιστικές με τη χαμηλότερη εκτιμώμενη τηλεοπτική αξία. Από αυτές αποδεσμεύει τυχαία δύο έως τέσσερις, κατευθύνοντας την αναζήτηση προς τις ασθενέστερες αγωνιστικές του προγράμματος.

- **Random Teams:** Επιλέγει δύο τυχαίες ομάδες και αποδεσμεύει όλες τις αναμετρήσεις στις οποίες συμμετέχει τουλάχιστον μία από αυτές, επιτρέποντας την αναδιάρθρωση του προγράμματός τους σε ολόκληρη τη διοργάνωση.

Οι τρεις operators δημιουργούν διαφορετικές γειτονιές αναζήτησης, συνδυάζοντας την τυχαία εξερεύνηση με τη στοχευμένη βελτίωση συγκεκριμένων τμημάτων του αγωνιστικού προγράμματος.


**CP-SAT Repair.** Μετά την επιλογή του destroy operator, πραγματοποιείται ανακατασκευή του αντίστοιχου τμήματος μέσω CP-SAT. Για τον σκοπό αυτό δημιουργείται αντίγραφο του αρχικού μοντέλου, στο οποίο οι αποφάσεις εκτός της επιλεγμένης γειτονιάς διατηρούνται σταθερές, ενώ οι υπόλοιπες επαναβελτιστοποιούνται. Επιπλέον, επιβάλλεται περιορισμός που αποκλείει την ακριβή αναπαραγωγή της τρέχουσας λύσης. Ο CP-SAT αναζητά μια νέα εφικτή ανάθεση, μεγιστοποιώντας την αντικειμενική συνάρτηση εντός του διαθέσιμου χρονικού ορίου. Εάν δεν εντοπιστεί κατάλληλη ανακατασκευή, η τρέχουσα λύση παραμένει αμετάβλητη.

**Adaptive Operator Selection.** Η επιλογή των destroy operators πραγματοποιείται με βάση προσαρμοστικά βάρη, τα οποία καθορίζουν τη σχετική πιθανότητα χρησιμοποίησης κάθε μηχανισμού. Αρχικά, οι τρεις operators διαθέτουν ίσα βάρη. Κατά τη διάρκεια της αναζήτησης, κάθε operator συγκεντρώνει ανταμοιβές ανάλογα με το αποτέλεσμα των επαναλήψεών του, με μεγαλύτερη ανταμοιβή όταν εντοπίζεται νέα καλύτερη λύση και μικρότερη όταν γίνεται αποδεκτή μια λύση χωρίς βελτίωση. Τα βάρη ανανεώνονται ανά 20 επαναλήψεις, συνδυάζοντας την προηγούμενη τιμή τους με τη μέση πρόσφατη απόδοσή τους. Έτσι, οι αποτελεσματικότεροι operators αποκτούν μεγαλύτερη πιθανότητα επιλογής, χωρίς να αποκλείεται η χρήση των υπολοίπων.

**Simulated Annealing.** Η αποδοχή των υποψήφιων λύσεων βασίζεται σε μηχανισμό Simulated Annealing, ώστε η αναζήτηση να μη περιορίζεται αποκλειστικά σε βελτιωτικές κινήσεις. Λύσεις με ίση ή μεγαλύτερη αντικειμενική τιμή γίνονται πάντοτε αποδεκτές, ενώ χειρότερες λύσεις μπορούν να γίνουν αποδεκτές με πιθανότητα που εξαρτάται από το μέγεθος της επιδείνωσης και την τρέχουσα θερμοκρασία. Πριν από την κύρια αναζήτηση πραγματοποιείται σύντομη φάση βαθμονόμησης, κατά την οποία καταγράφονται ενδεικτικές επιδεινώσεις για τον προσδιορισμό της αρχικής και της τελικής θερμοκρασίας. Στη συνέχεια, η θερμοκρασία μειώνεται σταδιακά με βάση τον διαθέσιμο χρόνο, περιορίζοντας προοδευτικά την πιθανότητα αποδοχής χειρότερων λύσεων.

**Διατήρηση Καλύτερης Λύσης και Τερματισμός.** Καθ' όλη τη διάρκεια της διαδικασίας διατηρούνται ανεξάρτητα η τρέχουσα λύση, από την οποία παράγονται οι επόμενες γειτονιές, και η καλύτερη εφικτή λύση που έχει εντοπιστεί. Η διάκριση αυτή επιτρέπει την προσωρινή αποδοχή χειρότερων λύσεων για την εξερεύνηση διαφορετικών περιοχών του χώρου αναζήτησης, χωρίς να χάνεται το καλύτερο αποτέλεσμα. Ο ALNS ολοκληρώνεται με την εξάντληση του προκαθορισμένου χρονικού προϋπολογισμού και επιστρέφει την καλύτερη λύση που εντοπίστηκε, χωρίς να προϋποθέτει απόδειξη καθολικής βελτιστότητας.



### 3.4. Επαλήθευση Εφικτότητας και Αξιολόγηση Ποιότητας Λύσεων

Η αξιολόγηση της προτεινόμενης μεθόδου περιλαμβάνει την ανεξάρτητη επαλήθευση των παραγόμενων προγραμμάτων, τη μέτρηση της βελτίωσης που επιτυγχάνει ο ALNS και την εκτίμηση της απόστασης των λύσεων από τη βέλτιστη δυνατή τιμή. Ιδιαίτερη έμφαση δίνεται στον υπολογισμό μαθηματικών άνω φραγμάτων (upper bounds), ώστε η ποιότητα των αποτελεσμάτων να αξιολογείται με κριτήρια πέρα από την αντικειμενική τιμή που επιτυγχάνει ο αλγόριθμος.

**Ανεξάρτητη Επαλήθευση.** Η υλοποίηση διαθέτει ανεξάρτητο μηχανισμό ελέγχου, ο οποίος επαληθεύει ότι κάθε παραγόμενο πρόγραμμα ικανοποιεί το σύνολο των υποχρεωτικών περιορισμών του Κεφαλαίου 2. Ο έλεγχος καλύπτει τη δομή του πρωταθλήματος, τη συμμετοχή των ομάδων, την κατανομή των αναμετρήσεων και των εδρών, την ανάθεση χρονικών ζωνών και τους πρόσθετους περιορισμούς ασφαλείας και προγραμματισμού. Εφαρμόζεται τόσο στην αρχική όσο και στην τελική λύση, καθώς και στις υποψήφιες λύσεις που παράγονται κατά τις επαναλήψεις του ALNS.

**Αξιολόγηση Αντικειμενικής Τιμής.** Για κάθε εφικτό πρόγραμμα επανυπολογίζονται ανεξάρτητα η συνολική εκτιμώμενη τηλεοπτική αξία, η αξία κάθε αγωνιστικής, η διαφορά μεταξύ της ισχυρότερης και της ασθενέστερης αγωνιστικής και η τελική αντικειμενική τιμή. Η αποτελεσματικότητα του ALNS αξιολογείται μέσω της σύγκρισης της αρχικής λύσης του CP-SAT με την καλύτερη λύση που εντοπίστηκε. Οι υπολογισμοί λαμβάνουν υπόψη τη διάκριση μεταξύ της μη στρογγυλοποιημένης αντικειμενικής συνάρτησης και της ακέραιας κλιμακωμένης μορφής που χρησιμοποιείται από τον επιλυτή.

**Άνω Φράγματα και Optimality Gap.** Η αξιολόγηση συμπληρώνεται με τον υπολογισμό έγκυρων άνω φραγμάτων, τα οποία επιτρέπουν την εκτίμηση της μέγιστης δυνατής αντικειμενικής τιμής. Για κάθε εξεταζόμενη τιμή της παραμέτρου ισορροπίας $\lambda$, το αντίστοιχο upper bound συγκρίνεται με την καλύτερη εφικτή λύση, παρέχοντας ένα σχετικό optimality gap. Με τον τρόπο αυτό διακρίνεται η βελτίωση που επιτυγχάνει ο αλγόριθμος από την ποιότητα που μπορεί να πιστοποιηθεί μαθηματικά.

**Αναλυτική Κατασκευή Άνω Φράγματος.**

Αρχικά, για κάθε αγωνιστική $r$, ταξινομούμε τις ελκυστικότητες των 14 ομάδων κατά φθίνουσα σειρά και τις αντιστοιχίζουμε στα επτά υψηλότερα διαθέσιμα βάρη χρονικών ζωνών, με κάθε βάρος να εμφανίζεται δύο φορές, μία για κάθε συμμετέχουσα ομάδα ενός αγώνα. Από την ανισότητα αναδιάταξης (Rearrangement Inequality) προκύπτει ένα άνω φράγμα $b_r$ της τηλεοπτικής αξίας κάθε αγωνιστικής, καθώς η αντιστοίχιση των μεγαλύτερων ελκυστικοτήτων με τα μεγαλύτερα βάρη μεγιστοποιεί το αντίστοιχο άθροισμα. Η κατασκευή αυτή αγνοεί ορισμένους περιορισμούς του αρχικού προβλήματος και επομένως δεν υποεκτιμά τη μέγιστη εφικτή αξία της αγωνιστικής.

Για την ενσωμάτωση της ποινής ισορροπίας εξετάζουμε, ενδεικτικά, την περίπτωση $\lambda=5$. Έστω $Q$ ένα σύνολο πέντε αγωνιστικών και $U_{\max}$, $U_{\min}$ η μέγιστη και η ελάχιστη τηλεοπτική αξία μεταξύ των 26 αγωνιστικών.

Εφόσον καμία αγωνιστική δεν μπορεί να υπερβαίνει το $U_{\max}$, ισχύει:

$$
\sum_{r\in Q}U_r\leq5U_{\max}
$$

Από την αντικειμενική συνάρτηση προκύπτει:

$$
\begin{aligned}
Z_5
&=\sum_{r=1}^{26}U_r-5(U_{\max}-U_{\min})\\
&=\sum_{r\notin Q}U_r+
\sum_{r\in Q}U_r-5U_{\max}+5U_{\min}\\
&\leq\sum_{r\notin Q}U_r+5U_{\min}\\
&\leq\sum_{r\notin Q}b_r+5b_{\min}
\end{aligned}
$$

όπου $b_{\min}$ είναι το μικρότερο από τα επιμέρους άνω φράγματα των αγωνιστικών.

Επιλέγοντας ως $Q$ τις πέντε αγωνιστικές με τα μεγαλύτερα $b_r$, προκύπτει το αναλυτικό άνω φράγμα:

$$
\boxed{
UB_{\mathrm{analytical}}(5)
=
\sum_{r\notin Q}b_r+5b_{\min}
}
$$

Η ίδια διαδικασία γενικεύεται για οποιαδήποτε ακέραια τιμή $\lambda$ μεταξύ 1 και 26, επιλέγοντας αντίστοιχα τις $\lambda$ αγωνιστικές με τα μεγαλύτερα επιμέρους άνω φράγματα. Για $\lambda=0$, το αναλυτικό upper bound προκύπτει από το άθροισμα όλων των $b_r$.

Τα αναλυτικά όρια μπορούν να ενισχυθούν μέσω χαλαρώσεων γραμμικού προγραμματισμού και πρόσθετων έγκυρων ανισοτήτων, οι οποίες αξιοποιούν μεγαλύτερο μέρος της δομής του προβλήματος. Για την πιστοποίηση αριθμητικών upper bounds της ακέραιας αντικειμενικής συνάρτησης απαιτείται επιπλέον συνεπής αντιμετώπιση της κλιμάκωσης και των στρογγυλοποιήσεων των συντελεστών.

**Ανάλυση Διαφορετικών Τιμών της Παραμέτρου $\lambda$.** Η παραπάνω μεθοδολογία επιτρέπει την αξιολόγηση του προβλήματος για διαφορετικές βαρύτητες της ποινής ισορροπίας. Για κάθε τιμή $\lambda$ θα εξετάζονται η καλύτερη αντικειμενική τιμή του ALNS, η συνολική τηλεοπτική αξία, η διαφορά μέγιστης και ελάχιστης αγωνιστικής, το αντίστοιχο έγκυρο upper bound και το σχετικό optimality gap. Η ανάλυση αυτή επιτρέπει τη διερεύνηση του συμβιβασμού μεταξύ της μεγιστοποίησης της συνολικής τηλεοπτικής αξίας και της ομοιόμορφης κατανομής της στις αγωνιστικές.

Τα αριθμητικά αποτελέσματα των παραπάνω ελέγχων και οι συγκρίσεις μεταξύ διαφορετικών τιμών της παραμέτρου παρουσιάζονται στο πειραματικό μέρος της εργασίας.
