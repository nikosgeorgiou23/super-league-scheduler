# 1. Εισαγωγή

### 1.1. Αθλητικός Προγραμματισμός και Τηλεοπτικές Απαιτήσεις

Ο προγραμματισμός των αγώνων ενός επαγγελματικού πρωταθλήματος ποδοσφαίρου αποτελεί ένα σύνθετο πρόβλημα συνδυαστικής βελτιστοποίησης, καθώς απαιτεί τον συντονισμό πολλών, συχνά αντικρουόμενων, απαιτήσεων. Η διαδικασία δεν περιορίζεται στον καθορισμό των αναμετρήσεων κάθε αγωνιστικής, αλλά περιλαμβάνει επίσης την ανάθεση των αγώνων σε συγκεκριμένες ημερομηνίες και χρονικές ζώνες, λαμβάνοντας υπόψη αγωνιστικούς, οργανωτικούς και εμπορικούς παράγοντες. Ιδιαίτερη σημασία έχουν οι τηλεοπτικές απαιτήσεις, καθώς η επιλογή της ημέρας και της ώρας διεξαγωγής ενός αγώνα μπορεί να επηρεάσει τη δυνητική τηλεθέαση και την εμπορική αξία της μετάδοσής του. Οι προτιμήσεις των τηλεοπτικών παρόχων ενδέχεται να εξαρτώνται από τη δημοτικότητα των ομάδων, την ελκυστικότητα των αναμετρήσεων και την κατανομή των αγώνων στις διαθέσιμες χρονικές ζώνες. Η ανάγκη ταυτόχρονης ικανοποίησης υποχρεωτικών περιορισμών και επιδίωξης πολλαπλών, ενδεχομένως ανταγωνιστικών, στόχων καθιστά τον προγραμματισμό ενός ποδοσφαιρικού πρωταθλήματος ιδιαίτερα απαιτητικό, αναδεικνύοντας τη δυνατότητα των μεθόδων μαθηματικής βελτιστοποίησης να προσφέρουν συστηματικές και τεκμηριωμένες λύσεις.

### 1.2. Προκλήσεις Προγραμματισμού στην Ελληνική Super League

Η ελληνική Super League παρουσιάζει ιδιαίτερες προκλήσεις στον προγραμματισμό των αγώνων, καθώς τα ζητήματα δημόσιας ασφάλειας και τα περιστατικά οπαδικής βίας εισάγουν πρόσθετη πολυπλοκότητα, πέρα από τις αγωνιστικές, οργανωτικές και εμπορικές απαιτήσεις που συναντώνται συνήθως στις επαγγελματικές ποδοσφαιρικές διοργανώσεις. Οι έντονες αντιπαλότητες μεταξύ συγκεκριμένων συλλόγων, τα προηγούμενα περιστατικά βίας και η πιθανότητα συγκρούσεων μεταξύ ομάδων οπαδών δημιουργούν αυξημένες απαιτήσεις για τη διαχείριση αναμετρήσεων υψηλής επικινδυνότητας. Ιδιαίτερη προσοχή πρέπει να δίνεται στη γεωγραφική εγγύτητα των γηπέδων, στον ταυτόχρονο προγραμματισμό αγώνων που ενδέχεται να προκαλέσουν ζητήματα ασφάλειας, καθώς και στη διαθεσιμότητα αστυνομικών δυνάμεων για την αποτελεσματική επιτήρηση των αθλητικών εκδηλώσεων. Η ενσωμάτωση τέτοιων απαιτήσεων σε ένα μαθηματικό μοντέλο βελτιστοποίησης μπορεί, επομένως, να συμβάλει στη δημιουργία προγραμμάτων που λαμβάνουν συστηματικά υπόψη τους παράγοντες που σχετίζονται με την ασφάλεια.

### 1.3. Συνδυαστική και Υπολογιστική Πολυπλοκότητα

Η μαθηματική δυσκολία του προγραμματισμού ενός ποδοσφαιρικού πρωταθλήματος προκύπτει τόσο από το μεγάλο πλήθος πιθανών χρονικών αναθέσεων όσο και από τις αλληλεξαρτήσεις μεταξύ των περιορισμών προγραμματισμού. Έστω $M$ το σύνολο των αγώνων μιας διοργάνωσης και $S_m$ το σύνολο των διαθέσιμων χρονικών ζωνών για κάθε αγώνα $m$. Υποθέτοντας ότι κάθε αγώνας μπορεί αρχικά να ανατεθεί ανεξάρτητα σε οποιαδήποτε από τις διαθέσιμες χρονικές ζώνες του, το συνολικό πλήθος των πιθανών αναθέσεων, πριν από την επιβολή πρόσθετων περιορισμών, δίνεται από τη σχέση:

$$
|\Omega_0| = \prod_{m \in M} |S_m|
$$

Ενδεικτικά, θεωρούμε ένα πρωτάθλημα που αποτελείται από 14 ομάδες, 26 αγωνιστικές και 182 αγώνες. Εάν κάθε αγώνας μπορεί να ανατεθεί σε μία από οκτώ διαθέσιμες χρονικές ζώνες, το θεωρητικό πλήθος των πιθανών αναθέσεων είναι:

$$
|\Omega_0| = 8^{182} \approx 2.30 \times 10^{164}
$$

Η τιμή αυτή αντιπροσωπεύει το μέγεθος του χώρου αναζήτησης χωρίς περιορισμούς και όχι το πλήθος των εφικτών προγραμμάτων.

Στην πράξη, οι αγωνιστικοί, οργανωτικοί, τηλεοπτικοί περιορισμοί, καθώς και οι απαιτήσεις δημόσιας ασφάλειας, δημιουργούν αλληλεξαρτήσεις που περιορίζουν σημαντικά τις διαθέσιμες αναθέσεις και καθιστούν πιο σύνθετη την αναζήτηση εφικτών λύσεων. Επιπλέον, ο εντοπισμός ενός εφικτού προγράμματος δεν εγγυάται απαραίτητα και τη βέλτιστη ποιότητά του, γεγονός που αναδεικνύει τη σημασία της αξιοποίησης εξειδικευμένων μεθόδων μαθηματικής βελτιστοποίησης.

### 1.4. Αντικείμενο της Εργασίας και Ερευνητική Συνεισφορά

Στην πράξη, ο προγραμματισμός των επαγγελματικών ποδοσφαιρικών πρωταθλημάτων πραγματοποιείται συχνά σε διαδοχικά στάδια. Αρχικά καθορίζονται οι αναμετρήσεις και η κατανομή τους στις αγωνιστικές, ενώ η διαδικασία του αναλυτικού χρονικού προγραμματισμού, που περιλαμβάνει την ανάθεση των αγώνων σε συγκεκριμένες ημερομηνίες και ώρες έναρξης, μπορεί να πραγματοποιηθεί σε μεταγενέστερο στάδιο, ανάλογα με τις διαθέσιμες πληροφορίες και τις λειτουργικές απαιτήσεις. Η προσέγγιση αυτή, η οποία παρατηρείται και στην ελληνική Super League, επιτρέπει τη σταδιακή λήψη αποφάσεων προγραμματισμού και την προσαρμογή τους σε μεταβαλλόμενες συνθήκες, όπως οι υποχρεώσεις των ομάδων στις ευρωπαϊκές διοργανώσεις, οι τηλεοπτικές προτεραιότητες και οι οργανωτικοί περιορισμοί.

Αντίθετα, η παρούσα εργασία εξετάζει μια προσέγγιση χρονικού προγραμματισμού σε επίπεδο ολόκληρης της αγωνιστικής περιόδου, στην οποία οι αναμετρήσεις και η κατανομή τους στις αγωνιστικές θεωρούνται προκαθορισμένες, ενώ οι ημερομηνίες και οι ώρες έναρξης όλων των αγώνων της κανονικής διάρκειας του πρωταθλήματος βελτιστοποιούνται στο πλαίσιο ενός ενιαίου μαθηματικού μοντέλου. Με τον τρόπο αυτό, καθίσταται δυνατή η ταυτόχρονη εξέταση απαιτήσεων προγραμματισμού που εκτείνονται σε ολόκληρη την αγωνιστική περίοδο, αντί της ανεξάρτητης αντιμετώπισής τους για κάθε αγωνιστική.

Πρωταρχικός στόχος είναι η κατασκευή ενός εφικτού προγράμματος που ικανοποιεί όλους τους υποχρεωτικούς περιορισμούς, επιδιώκοντας παράλληλα την ισορροπημένη κατανομή της τηλεοπτικής ελκυστικότητας μεταξύ των αγωνιστικών. Μια τέτοια προσέγγιση είναι ιδιαίτερα κατάλληλη για διοργανώσεις με σχετικά σταθερό αγωνιστικό ημερολόγιο και περιορισμένες εξωτερικές υποχρεώσεις, συμπεριλαμβανομένων πρωταθλημάτων των οποίων οι συμμετέχουσες ομάδες δεν αγωνίζονται σε διεθνείς διασυλλογικές διοργανώσεις.

Το προτεινόμενο μοντέλο αποσκοπεί στην ανάπτυξη μιας διαφανούς και συστηματικής μεθοδολογίας για τον χρονικό προγραμματισμό ολόκληρης της αγωνιστικής περιόδου, με δυνατότητα μελλοντικής προσαρμογής σε πιο σύνθετα αγωνιστικά περιβάλλοντα. Η υλοποίησή του βασίζεται σε ρητά καθορισμένες παραδοχές σχετικά με τη διαθεσιμότητα των δεδομένων και τις συνθήκες διεξαγωγής του πρωταθλήματος, οι οποίες παρουσιάζονται στο επόμενο κεφάλαιο, όπου διατυπώνεται αναλυτικά το πρόβλημα.


# 2. Διατύπωση του Προβλήματος

### 2.1. Δομή του Πρωταθλήματος

- Στο πρωτάθλημα συμμετέχουν 14 ομάδες και η κανονική διάρκεια αποτελείται από δύο γύρους των 13 αγωνιστικών:
  - **Πρώτος γύρος:** Αγωνιστικές 1–13.
  - **Δεύτερος γύρος:** Αγωνιστικές 14–26.


**Μαθηματική Διατύπωση**

Έστω $T$ το σύνολο των ομάδων και $R$ το σύνολο των αγωνιστικών.

$$
|T|=14
$$

$$
R=\{1,2,\ldots,26\}
$$

Ορίζουμε τα σύνολα αγωνιστικών των δύο γύρων:

$$
R_1=\{1,2,\ldots,13\}
$$

$$
R_2=\{14,15,\ldots,26\}
$$

με:

$$
R=R_1\cup R_2,\qquad R_1\cap R_2=\varnothing
$$

Το σύνολο $T$ περιλαμβάνει τις ακόλουθες 14 ομάδες:

- Παναθηναϊκός
- Ολυμπιακός
- ΑΕΚ
- ΠΑΟΚ
- Άρης
- Ηρακλής
- Ατρόμητος
- Κηφισιά
- Καλαμάτα
- Αστέρας Τρίπολης
- Παναιτωλικός
- ΟΦΗ
- Λεβαδειακός
- Βόλος

Επιπλέον, ορίζουμε το σύνολο $M$ των αναμετρήσεων του πρωταθλήματος:

$$
M=\{(i,j)\in T\times T\mid i\neq j\}
$$

Κάθε διατεταγμένο ζεύγος $(i,j)\in M$ αντιπροσωπεύει μία αναμέτρηση στην οποία η ομάδα $i$ είναι γηπεδούχος και η ομάδα $j$ φιλοξενούμενη. Επομένως, οι αναμετρήσεις $(i,j)$ και $(j,i)$ θεωρούνται διαφορετικές, καθώς αντιστοιχούν σε αγώνες με αντίστροφες έδρες.



- Κάθε ομάδα συμμετέχει σε ακριβώς μία αναμέτρηση ανά αγωνιστική, είτε ως γηπεδούχος είτε ως φιλοξενούμενη.

**Μαθηματική Διατύπωση**

Ορίζουμε τη δυαδική μεταβλητή απόφασης:

$$
x_{ijr}=
\begin{cases}
1, & \text{αν η ομάδα }i\text{ υποδέχεται την }j\text{ στην αγωνιστική }r\\
0, & \text{διαφορετικά}
\end{cases}
$$

για $i,j\in T$, $i\neq j$ και $r\in R$.

Ο περιορισμός συμμετοχής κάθε ομάδας είναι:

$$
\sum_{\substack{j\in T\\j\neq i}}
(x_{ijr}+x_{jir})=1
\qquad \forall i\in T,\ \forall r\in R
$$



- Κάθε ζευγάρι ομάδων αναμετράται ακριβώς μία φορά σε κάθε γύρο, με αντίστροφες έδρες μεταξύ πρώτου και δεύτερου γύρου. Η σειρά των αναμετρήσεων στον δεύτερο γύρο δεν είναι υποχρεωτικά ίδια με εκείνη του πρώτου.

**Μαθηματική Διατύπωση**

Για κάθε ζευγάρι διαφορετικών ομάδων $i,j\in T$, με $i<j$, επιβάλλεται μία αναμέτρηση ανά γύρο:

$$
\sum_{r\in R_1}(x_{ijr}+x_{jir})=1
\qquad \forall i,j\in T,\ i<j
$$

$$
\sum_{r\in R_2}(x_{ijr}+x_{jir})=1
\qquad \forall i,j\in T,\ i<j
$$

Για να εξασφαλιστεί η αντιστροφή των εδρών μεταξύ των δύο γύρων:

$$
\sum_{r\in R_1}x_{ijr}
=
\sum_{r\in R_2}x_{jir}
\qquad \forall i,j\in T,\ i\neq j
$$



**Παράγωγες Ιδιότητες**

Από τους παραπάνω περιορισμούς προκύπτουν οι ακόλουθες ιδιότητες:

- Κάθε ομάδα δίνει συνολικά 13 εντός έδρας και 13 εκτός έδρας αγώνες.
- Κάθε αγωνιστική περιλαμβάνει ακριβώς 7 αναμετρήσεις.
- Στην κανονική διάρκεια του πρωταθλήματος διεξάγονται συνολικά 182 αγώνες.

**Μαθηματική Διατύπωση**

Το πλήθος των εντός και εκτός έδρας αναμετρήσεων κάθε ομάδας είναι:

$$
H_i=A_i=|T|-1=13,\qquad \forall i\in T
$$

όπου $H_i$ και $A_i$ συμβολίζουν το πλήθος των εντός και εκτός έδρας αγώνων της ομάδας $i$, αντίστοιχα.

Το πλήθος των αναμετρήσεων ανά αγωνιστική είναι:

$$
\frac{|T|}{2}=\frac{14}{2}=7
$$

Το συνολικό πλήθος των αναμετρήσεων του πρωταθλήματος είναι:

$$
|M|=|T|(|T|-1)=14\cdot13=182
$$

Οι παραπάνω σχέσεις αποτελούν συνέπειες των βασικών περιορισμών και δεν απαιτείται να επιβληθούν ως πρόσθετοι ανεξάρτητοι περιορισμοί στο μαθηματικό μοντέλο.



### 2.2. Χαρακτηριστικά Ομάδων

**Κατηγοριοποίηση Ομάδων Υψηλού Κινδύνου**

- Για τις ανάγκες του προγραμματισμού, ορισμένες ομάδες χαρακτηρίζονται ως υψηλού κινδύνου και διαχωρίζονται σε δύο κατηγορίες με βάση τη γεωγραφική τους τοποθεσία:
  - **Ομάδες υψηλού κινδύνου Αθήνας:** Παναθηναϊκός, Ολυμπιακός, ΑΕΚ.
  - **Ομάδες υψηλού κινδύνου Θεσσαλονίκης:** ΠΑΟΚ, Άρης, Ηρακλής.
- Οι υπόλοιπες οκτώ ομάδες δεν εντάσσονται σε κάποια από τις δύο κατηγορίες.
- Η κατηγοριοποίηση αυτή χρησιμοποιείται για τον προσδιορισμό των αναμετρήσεων υψηλής επικινδυνότητας και την εφαρμογή πρόσθετων περιορισμών ασφαλείας κατά τον χρονικό προγραμματισμό.

**Μαθηματική Διατύπωση**

Ορίζουμε τα ακόλουθα υποσύνολα του συνόλου ομάδων $T$:

$$
T_A=\{\text{Παναθηναϊκός, Ολυμπιακός, ΑΕΚ}\}
$$

$$
T_\Theta=\{\text{ΠΑΟΚ, Άρης, Ηρακλής}\}
$$

όπου $T_A$ και $T_\Theta$ αντιπροσωπεύουν τις ομάδες υψηλού κινδύνου Αθήνας και Θεσσαλονίκης, αντίστοιχα.

Τα δύο σύνολα είναι ξένα μεταξύ τους:

$$
T_A\cap T_\Theta=\varnothing
$$

Το σύνολο των υπόλοιπων ομάδων ορίζεται ως:

$$
T_O=T\setminus(T_A\cup T_\Theta)
$$

με:

$$
|T_A|=3,\qquad |T_\Theta|=3,\qquad |T_O|=8
$$




**Δυναμικότητα Ομάδων**

Για την εκτίμηση της αγωνιστικής ισχύος κάθε ομάδας εισάγεται ένας αριθμητικός δείκτης δυναμικότητας, ο οποίος συνδυάζει δύο βασικούς παράγοντες:

- **Την ιστορική αγωνιστική επίδοση**, όπως αποτυπώνεται από την τελική κατάταξη της ομάδας στα τρία προηγούμενα πρωταθλήματα.
- **Την τρέχουσα χρηματιστηριακή αξία του ρόστερ**, η οποία χρησιμοποιείται ως προσεγγιστικός δείκτης της σημερινής αγωνιστικής και οικονομικής ισχύος.

Η κατασκευή του δείκτη πραγματοποιείται σταδιακά, ξεκινώντας από την αξιολόγηση της ιστορικής επίδοσης και ενσωματώνοντας στη συνέχεια την οικονομική διάσταση.

**1. Ιστορική Αγωνιστική Επίδοση**

- Η ιστορική επίδοση κάθε ομάδας αξιολογείται με βάση την τελική της κατάταξη στις τρεις προηγούμενες αγωνιστικές περιόδους (2023–24, 2024–25 και 2025–26).
- Οι θέσεις κατάταξης μετατρέπονται σε αριθμητική βαθμολογία, ώστε οι υψηλότερες θέσεις να αντιστοιχούν σε μεγαλύτερη δυναμικότητα.
- Οι πιο πρόσφατες αγωνιστικές περίοδοι λαμβάνουν μεγαλύτερη βαρύτητα, καθώς θεωρούνται περισσότερο αντιπροσωπευτικές της σημερινής αγωνιστικής ισχύος.
- Σε περίπτωση που μια ομάδα δεν συμμετείχε στην πρώτη κατηγορία κατά κάποια από τις εξεταζόμενες περιόδους, της αποδίδεται συμβολικά η 15η θέση, δηλαδή μία θέση χαμηλότερα από την τελευταία του πρωταθλήματος.

**Μαθηματική Διατύπωση**

Έστω $p_{i,y}$ η τελική θέση της ομάδας $i$ στην αγωνιστική περίοδο $y$.

Ορίζουμε τη συνάρτηση βαθμολόγησης της ιστορικής κατάταξης:

$$
P(p)=100\frac{14-p}{13}
$$

Η συνάρτηση αυτή αντιστοιχίζει την πρώτη θέση σε 100 μονάδες και τη 14η θέση σε 0 μονάδες:

$$
P(1)=100,\qquad P(14)=0
$$

Κάθε βελτίωση μίας θέσης αυξάνει τη βαθμολογία κατά:

$$
\Delta P=\frac{100}{13}\approx7.69
$$

Για τις ομάδες που δεν συμμετείχαν στην πρώτη κατηγορία κατά κάποια αγωνιστική περίοδο, ορίζεται:

$$
p_{i,y}=15
$$

Επομένως:

$$
P(15)=-\frac{100}{13}\approx-7.69
$$

Η αρνητική τιμή εκφράζει την παραδοχή ότι η απουσία από την πρώτη κατηγορία αξιολογείται χαμηλότερα από την κατάληψη της τελευταίας θέσης.

Στη συνέχεια, υπολογίζεται η σταθμισμένη ιστορική επίδοση της ομάδας $i$:

$$
H_i=
0.20P(p_{i,23/24})
+0.30P(p_{i,24/25})
+0.50P(p_{i,25/26})
$$

Οι συντελεστές $0.20$, $0.30$ και $0.50$ αποδίδουν αυξανόμενη βαρύτητα στις πιο πρόσφατες αγωνιστικές περιόδους, με την τελευταία να συνεισφέρει κατά 50% στη συνολική ιστορική αξιολόγηση.

**2. Ενσωμάτωση της Χρηματιστηριακής Αξίας**

Η ιστορική κατάταξη δεν επαρκεί από μόνη της για την αποτύπωση της σημερινής δυναμικότητας μιας ομάδας. Μεταβολές στη σύνθεση του ρόστερ, μεταγραφικές επενδύσεις και αλλαγές στην οικονομική κατάσταση ενός συλλόγου μπορούν να διαφοροποιήσουν σημαντικά την αγωνιστική του ισχύ σε σχέση με τις προηγούμενες περιόδους.

Για τον λόγο αυτό, στον δείκτη δυναμικότητας ενσωματώνεται η συνολική χρηματιστηριακή αξία του ρόστερ, όπως εκτιμάται από την πλατφόρμα Transfermarkt. Η αξία αυτή χρησιμοποιείται ως προσεγγιστικό μέτρο της τρέχουσας ισχύος μιας ομάδας και δεν ταυτίζεται με τον πραγματικό μεταγραφικό ή λειτουργικό προϋπολογισμό της.

Επιλέγεται λογαριθμικός μετασχηματισμός της χρηματιστηριακής αξίας, ώστε η συνεισφορά της να αυξάνεται με φθίνοντα ρυθμό. Με τον τρόπο αυτό, περιορίζεται η επίδραση πολύ μεγάλων οικονομικών διαφορών μεταξύ των ομάδων, χωρίς να εξαλείφεται το πλεονέκτημα των συλλόγων με υψηλότερη αποτίμηση.

**Μαθηματική Διατύπωση**

Έστω $B_i>0$ η συνολική χρηματιστηριακή αξία του ρόστερ της ομάδας $i$, εκφρασμένη σε εκατομμύρια ευρώ.

Ορίζουμε την οικονομική συνιστώσα της δυναμικότητας:

$$
E_i=7.5\log_2(B_i)
$$

Η χρήση λογαρίθμου με βάση το $2$ σημαίνει ότι κάθε διπλασιασμός της χρηματιστηριακής αξίας αυξάνει την οικονομική συνιστώσα κατά $7.5$ μονάδες:

$$
E_i(2B_i)-E_i(B_i)=7.5
$$

**3. Τελικός Δείκτης Δυναμικότητας**

Ο τελικός δείκτης δυναμικότητας προκύπτει από τον συνδυασμό της σταθμισμένης ιστορικής επίδοσης και της λογαριθμικής χρηματιστηριακής αξίας.

Η ιστορική συνιστώσα πολλαπλασιάζεται με τον συντελεστή $0.30$, ενώ η οικονομική συνιστώσα διατηρεί τον συντελεστή $7.5$. Οι συντελεστές αυτοί καθορίζουν τη σχετική επιρροή των δύο παραγόντων στον τελικό δείκτη και αποτελούν παραδοχές της παρούσας εργασίας.

**Τελική Μαθηματική Διατύπωση**

Για κάθε ομάδα $i\in T$, ο δείκτης δυναμικότητας ορίζεται ως:

$$
\boxed{
F_i=0.30H_i+7.5\log_2(B_i)
}
$$

όπου:

$$
H_i=
0.20P(p_{i,23/24})
+0.30P(p_{i,24/25})
+0.50P(p_{i,25/26})
$$

και:

$$
P(p)=100\frac{14-p}{13}
$$

Για την καλύτερη κατανόηση της επίδρασης των συντελεστών:

- Κάθε διπλασιασμός της χρηματιστηριακής αξίας του ρόστερ αυξάνει τον τελικό δείκτη κατά **7.5 μονάδες**.
- Μία θέση καλύτερη στην κατάταξη της περιόδου 2025–26, χωρίς μεταβολή των άλλων παραγόντων, αυξάνει τον τελικό δείκτη κατά περίπου **1.15 μονάδες**.
- Μία θέση καλύτερη και στις τρεις εξεταζόμενες περιόδους αυξάνει τον τελικό δείκτη κατά περίπου **2.31 μονάδες**.

Ο δείκτης $F_i$ αποτελεί μια προσεγγιστική αποτύπωση της δυναμικότητας κάθε ομάδας, βασισμένη σε ιστορικά αγωνιστικά δεδομένα και στην τρέχουσα αποτίμηση του ρόστερ της. Οι συντελεστές του αποτελούν επιλογές μοντελοποίησης και όχι αποτελέσματα στατιστικής εκτίμησης.

Η δυναμικότητα αυτή χρησιμοποιείται στη συνέχεια για τον προσδιορισμό της εκτιμώμενης τηλεοπτικής ελκυστικότητας των αναμετρήσεων αλλά και την αξία των τηλεοπτικών παραθύρων.
